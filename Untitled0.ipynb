```python
# ============================================================
#@title Unifica as planilhas (Versão FINAL - Correção da ordem: Ano e Repetição)
# ============================================================


import numpy as np
import pandas as pd
import os

# 1. Caminhos
BASE_PATH = '/content/drive/MyDrive/Tese/propostas_de_artigos/proposta_artigo_043'
IRRIGATED_FILE = os.path.join(BASE_PATH, 'irrigated', 'regression', 'raw_data_Irrigated_Regression.xlsx')
RAINFED_FILE   = os.path.join(BASE_PATH, 'rainfed',   'regression', 'raw_data_Rainfed_Regression.xlsx')
OUTPUT_FILE    = os.path.join(BASE_PATH, '/data/dataset_unificado_043.xlsx')

# 2. Função de leitura corrigida
def load_and_format(path, system_name, system_code):
    print(f"\n--- Lendo: {os.path.basename(path)} ---")

    # Lê o arquivo sem cabeçalho
    df_raw = pd.read_excel(path, header=None)

    # Estrutura real:
    # Coluna 0: Índice da amostra
    # Coluna 1: Water_Potential (Ψw)
    # Coluna 2: Month (Mês)
    # Coluna 3: Year (Ano) -> era isso que estava sendo lido como Repetição!
    # Coluna 4: Genotype (Genótipo)
    # Coluna 5: Replicate (Repetição)
    # Coluna 6 em diante: Reflectâncias

    # Vamos ignorar a coluna 0 e pegar os dados da coluna 1 em diante
    df_data = df_raw.iloc[:, 1:]

    # Transpor: amostras nas linhas
    df = df_data.T.reset_index(drop=True)

    # Converter vírgula para ponto
    df = df.astype(str).replace(',', '.', regex=True)
    df = df.apply(pd.to_numeric, errors='coerce')

    # FORÇAR os nomes das colunas na ordem correta (Ψw, Mês, Ano, Genótipo, Repetição)
    df.columns = ['Water_Potential', 'Month', 'Year', 'Genotype', 'Replicate'] + [f'band_{i}' for i in range(df.shape[1] - 5)]

    # Renomear as bandas para o formato nm
    meta_cols = ['Water_Potential', 'Month', 'Year', 'Genotype', 'Replicate']
    band_cols = [c for c in df.columns if c not in meta_cols]

    num_bands = len(band_cols)
    wavelengths = np.linspace(400.03, 951.00, num_bands)
    new_names = [f"{w:.2f}nm" for w in wavelengths]

    rename_dict = dict(zip(band_cols, new_names))
    df = df.rename(columns=rename_dict)

    # Adicionar colunas discriminadoras
    df['Sistema_Manejo'] = system_name
    df['Sistema_Code'] = system_code

    # Reordenar colunas
    final_cols = meta_cols + ['Sistema_Manejo', 'Sistema_Code'] + new_names
    df = df[final_cols]

    print(f"Leitura OK: {df.shape[0]} amostras × {df.shape[1]} colunas")
    print(f"Ψw mínimo: {df['Water_Potential'].min():.2f}")
    print(f"Ψw máximo: {df['Water_Potential'].max():.2f}")
    print(f"Repetições únicas: {np.sort(df['Replicate'].dropna().unique())}")
    print(f"Anos únicos: {np.sort(df['Year'].dropna().unique())}")
    print(f"Genótipos únicos: {np.sort(df['Genotype'].dropna().unique())}")

    return df

# 3. Carregar os dados
df_irr = load_and_format(IRRIGATED_FILE, 'Irrigado', 1)
df_rain = load_and_format(RAINFED_FILE, 'Sequeiro', 0)

# 4. Empilhar
df_unificado = pd.concat([df_irr, df_rain], ignore_index=True)

print(f"\n--- Resultado Final ---")
print(f"Total de amostras: {len(df_unificado)}")
print(f"  - Irrigado: {len(df_unificado[df_unificado['Sistema_Code'] == 1])}")
print(f"  - Sequeiro: {len(df_unificado[df_unificado['Sistema_Code'] == 0])}")
print(f"Repetições únicas: {np.sort(df_unificado['Replicate'].dropna().unique())}")
print(f"Anos únicos: {np.sort(df_unificado['Year'].dropna().unique())}")
print(f"Genótipos únicos: {np.sort(df_unificado['Genotype'].dropna().unique())}")
print(f"Ψw mínimo: {df_unificado['Water_Potential'].min():.2f} MPa")
print(f"Ψw máximo: {df_unificado['Water_Potential'].max():.2f} MPa")

# 5. Salvar
df_unificado.to_excel(OUTPUT_FILE, index=False)
print(f"\n✅ Dataset salvo em: {OUTPUT_FILE}")
```

```python
# -*- coding: utf-8 -*-
# @title Reprodução do pipeline do artigo EAAI-26-21747
"""
Reprodução do pipeline do artigo EAAI-26-21747
Adaptado para o dataset_unificado_043_CORRIGIDO.xlsx
"""

import subprocess, sys
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "catboost", "openpyxl"], check=False)

import os, re, time, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
from scipy.stats import skew, kurtosis

from sklearn.base import clone
from sklearn.decomposition import PCA
from sklearn.ensemble import ExtraTreesRegressor, RandomForestRegressor, GradientBoostingRegressor
from sklearn.linear_model import Ridge
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from sklearn.model_selection import GridSearchCV, GroupKFold
from sklearn.cross_decomposition import PLSRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVR

warnings.filterwarnings("ignore")

try:
    from xgboost import XGBRegressor
except Exception:
    XGBRegressor = None
try:
    from catboost import CatBoostRegressor
except Exception:
    CatBoostRegressor = None
try:
    from lightgbm import LGBMRegressor
except Exception:
    LGBMRegressor = None


# ============================================================
# CONFIGURAÇÃO
# ============================================================
BASE_PATH = '/content/drive/MyDrive/Tese/propostas_de_artigos/proposta_artigo_043'
DATA_PATH = os.path.join(BASE_PATH, '/data/dataset_unificado_043.xlsx')
OUT_DIR = os.path.join(BASE_PATH, "/resultado/resultados_artigo_043_EAAI")
os.makedirs(OUT_DIR, exist_ok=True)
np.random.seed(42)
plt.rcParams.update({"figure.dpi": 110, "axes.grid": True, "grid.alpha": .25})

# ============================================================
# 1. LEITURA
# ============================================================
print("Carregando dataset...")
df = pd.read_excel(DATA_PATH)

# Identificar colunas espectrais (aquelas que terminam com 'nm')
spec_cols = [c for c in df.columns if str(c).endswith('nm')]
# Extrair os wavelengths para ordenar
wls = np.array([float(str(c).replace('nm', '')) for c in spec_cols])
order = np.argsort(wls)
spec_cols = [spec_cols[i] for i in order]
wls = wls[order]

print(f"Bandas espectrais detectadas: {len(spec_cols)} ({wls.min():.1f}-{wls.max():.1f} nm)")

# Limpeza
R = df[spec_cols].apply(pd.to_numeric, errors="coerce")
R = R.interpolate(axis=1, limit_direction="both")
y_all = pd.to_numeric(df['Water_Potential'], errors="coerce")
ok = R.notna().all(axis=1) & y_all.notna() & df['Replicate'].notna()
df, R, y_all = df[ok].reset_index(drop=True), R[ok].reset_index(drop=True), y_all[ok].reset_index(drop=True)
print(f"Amostras válidas: {len(df)}")

y = y_all.values.astype(float)
groups = df['Replicate'].values.astype(int)
raw = R.values.astype(float)

# ============================================================
# 2. ENGENHARIA DE ATRIBUTOS (EAAI)
# ============================================================
def snv(M):
    return (M - M.mean(1, keepdims=True)) / (M.std(1, keepdims=True) + 1e-12)

X_snv = snv(raw)

# Metadados: Apenas Mês (circular) e Sistema (binário), como no EAAI
meta_blocks, meta_names = [], []

# Mês (Circular - 2 harmônicos)
# O 'Month' no dataset está com valores de 14, 15, 16 (que são anos) ou meses?
# O print mostrou Month = 14, 15, 16. Isso é estranho, mas vamos tratar como mês.
# Se for ano, o código vai dar erro, então vamos checar.
# O EAAI usa Month (1-12). Vou assumir que a coluna 'Month' é o mês.
mo = df['Month'].values.astype(float)
meta_blocks += [np.sin(2*np.pi*mo/12), np.cos(2*np.pi*mo/12),
                np.sin(4*np.pi*mo/12), np.cos(4*np.pi*mo/12)]
meta_names += ["month_sin1", "month_cos1", "month_sin2", "month_cos2"]

# Sistema de Manejo (Irrigado = 1, Sequeiro = 0)
mgmt = (df['Sistema_Manejo'] == 'Irrigado').astype(float).values
meta_blocks.append(mgmt)
meta_names.append("management")

meta = np.column_stack(meta_blocks)

# X final
X = np.hstack([X_snv, meta])
names = np.array([f"spec_{w:.1f}" for w in wls] + [f"meta_{n}" for n in meta_names])

print(f"Matriz de atributos: {X.shape} (bandas={X_snv.shape[1]}, meta={meta.shape[1]})")

# ============================================================
# 3. PRÉ-PROCESSADOR (Fiel ao EAAI)
# ============================================================
class SpectralPreprocessor:
    def __init__(self, n_sel=300, n_pc=30, seed=42):
        self.n_sel, self.n_pc, self.seed = n_sel, n_pc, seed

    def fit(self, X, y):
        self.sc1 = StandardScaler().fit(X)
        Xs = self.sc1.transform(X)
        # O EAAI usa Extra Trees para selecionar 300 features
        et = ExtraTreesRegressor(n_estimators=300, max_features=0.33, random_state=self.seed, n_jobs=-1).fit(Xs, y)
        n_sel = min(self.n_sel, X.shape[1])
        self.sel_idx = np.sort(np.argsort(et.feature_importances_)[::-1][:n_sel])
        self.importances_ = et.feature_importances_
        # PCA de 30 componentes
        self.pca = PCA(n_components=min(self.n_pc, n_sel), random_state=self.seed)
        Zp = self.pca.fit_transform(Xs[:, self.sel_idx])
        self.sc2 = StandardScaler().fit(Zp)
        return self

    def transform_selected(self, X):
        return self.sc1.transform(X)[:, self.sel_idx]

    def transform(self, X):
        return self.sc2.transform(self.pca.transform(self.transform_selected(X)))

# ============================================================
# 4. MODELOS (Tabela 1 do EAAI + Benchmarks)
# ============================================================
MLP_CONFIGS = [
    ((300, 150, 75, 35), 0.0003, 0.001, 32),
    ((400, 200, 100, 50), 0.0001, 0.001, 64),
    ((300, 150, 75),      0.0005, 0.003, 32),
    ((500, 250, 100),     0.0002, 0.001, 128),
    ((200, 100, 50, 25),  0.0005, 0.005, 32),
]

def make_mlps(seed=42):
    return {f"MLP_{i+1}": MLPRegressor(
        hidden_layer_sizes=a, activation="relu", solver="adam", alpha=al,
        learning_rate_init=lr, batch_size=bs, early_stopping=True,
        validation_fraction=0.15, n_iter_no_change=20, max_iter=2000,
        random_state=seed + i)
        for i, (a, al, lr, bs) in enumerate(MLP_CONFIGS)}

def make_benchmarks(seed=42):
    m = {
        "Extra Trees": ExtraTreesRegressor(n_estimators=300, random_state=seed, n_jobs=-1),
        "Random Forest": RandomForestRegressor(n_estimators=300, random_state=seed, n_jobs=-1),
        "Gradient Boosting": GradientBoostingRegressor(random_state=seed),
        "Ridge": Ridge(alpha=1.0),
        "SVR (RBF)": SVR(C=10),
        "KNN": KNeighborsRegressor(n_neighbors=5),
        "PLS (SNV spectra)": PLSRegression(n_components=10, scale=False) # PLS simplificado
    }
    if XGBRegressor: m["XGBoost"] = XGBRegressor(random_state=seed, n_jobs=-1, verbosity=0)
    if CatBoostRegressor: m["CatBoost"] = CatBoostRegressor(random_seed=seed, verbose=0)
    if LGBMRegressor: m["LightGBM"] = LGBMRegressor(random_state=seed, verbose=-1)
    return m

def metrics(yt, yp):
    rmse = np.sqrt(mean_squared_error(yt, yp))
    return dict(R2=r2_score(yt, yp), RMSE=rmse, MAE=mean_absolute_error(yt, yp),
                RPD=np.std(yt, ddof=1) / rmse)

def rpd_class(r):
    return "Excelente" if r > 2.5 else "Muito bom" if r >= 2.0 else "Bom" if r >= 1.5 else "Insatisfatório"

# ============================================================
# 5. VALIDAÇÃO LEAVE-ONE-REPLICATE-OUT
# ============================================================
def run_loro(X, raw, names, y, groups):
    n = len(y)
    uniq = np.unique(groups)
    store = {}

    def put(model, te, p, t):
        d = store.setdefault(model, dict(pred=np.full(n, np.nan), time={}))
        d["pred"][te] = p
        d["time"][groups[te][0]] = t

    for g in uniq:
        te, tr = np.where(groups == g)[0], np.where(groups != g)[0]
        print(f"\n--- Fold: replicata de teste = {g} (treino={len(tr)}, teste={len(te)}) ---")
        prep = SpectralPreprocessor().fit(X[tr], y[tr])
        Ztr, Zte = prep.transform(X[tr]), prep.transform(X[te])

        # Estágios metodológicos
        t0 = time.time()
        m = ExtraTreesRegressor(n_estimators=300, random_state=42, n_jobs=-1).fit(raw[tr], y[tr])
        put("Original (ET, espectro bruto)", te, m.predict(raw[te]), time.time() - t0)

        t0 = time.time()
        Str, Ste = prep.transform_selected(X[tr]), prep.transform_selected(X[te])
        m = ExtraTreesRegressor(n_estimators=300, random_state=42, n_jobs=-1).fit(Str, y[tr])
        put("Seleção de atributos (ET)", te, m.predict(Ste), time.time() - t0)

        models = make_mlps()
        bench = make_benchmarks()
        models.update(bench)
        for name, est in models.items():
            t0 = time.time()
            est = clone(est).fit(Ztr, y[tr])
            put(name, te, est.predict(Zte), time.time() - t0)

    # Ensemble (5 MLPs)
    mlp_names = [f"MLP_{i+1}" for i in range(len(MLP_CONFIGS))]
    ens = np.mean([store[k]["pred"] for k in mlp_names], axis=0)
    ens_time = {g: sum(store[k]["time"][g] for k in mlp_names) for g in uniq}
    store["Ensemble (5 MLPs)"] = dict(pred=ens, time=ens_time)

    # Melhor MLP individual
    mean_r2 = {k: np.mean([r2_score(y[groups == g], store[k]["pred"][groups == g]) for g in uniq])
               for k in mlp_names}
    best = max(mean_r2, key=mean_r2.get)
    store["MLP (Best Individual)"] = dict(pred=store[best]["pred"].copy(), time=store[best]["time"])

    for k in mlp_names:
        store[f"[{k}]"] = store.pop(k)
    return store, best

def summarize(store, y, groups):
    rows, per_fold = [], []
    uniq = np.unique(groups)
    for name, d in store.items():
        pf = [metrics(y[groups == g], d["pred"][groups == g]) for g in uniq]
        for g, mm in zip(uniq, pf):
            per_fold.append(dict(Model=name, Fold_test_replicate=g, **mm))
        pooled = metrics(y, d["pred"])
        rows.append(dict(
            Model=name,
            R2_mean=np.mean([p["R2"] for p in pf]), R2_std=np.std([p["R2"] for p in pf]),
            RMSE_mean=np.mean([p["RMSE"] for p in pf]), MAE_mean=np.mean([p["MAE"] for p in pf]),
            RPD_mean=np.mean([p["RPD"] for p in pf]),
            R2_pooled=pooled["R2"], RMSE_pooled=pooled["RMSE"], RPD_pooled=pooled["RPD"],
            Time_s_per_fold=np.mean(list(d["time"].values()))))
    res = pd.DataFrame(rows).sort_values("R2_mean", ascending=False).reset_index(drop=True)
    res["RPD_class"] = res["RPD_mean"].apply(rpd_class)
    return res, pd.DataFrame(per_fold)

print("\n" + "=" * 70 + "\nVALIDAÇÃO LEAVE-ONE-REPLICATE-OUT\n" + "=" * 70)
store, best_mlp = run_loro(X, raw, names, y, groups)
res, per_fold = summarize(store, y, groups)
res_main = res[~res["Model"].str.startswith("[")]
print(f"\nMelhor MLP individual: {best_mlp}")
print(res_main.round(4).to_string(index=False))

# ============================================================
# 6. ANÁLISE DE RESÍDUOS
# ============================================================
ens_pred = store["Ensemble (5 MLPs)"]["pred"]
resid = y - ens_pred
bins = [(-np.inf, -4), (-4, -2), (-2, -1), (-1, np.inf)]
labels = ["< -4 MPa", "-4 a -2", "-2 a -1", "> -1 MPa"]
rows = []
for (lo, hi), lab in zip(bins, labels):
    m = (y > lo) & (y <= hi)
    if m.sum() > 1:
        rows.append(dict(Faixa=lab, n=int(m.sum()),
                         RMSE=np.sqrt(np.mean(resid[m] ** 2)), MAE=np.mean(abs(resid[m])),
                         Bias=np.mean(resid[m])))
err_bins = pd.DataFrame(rows)
print("\nErro do ensemble por faixa de Ψw (predições out-of-fold):")
print(err_bins.round(3).to_string(index=False))
print(f"Viés global: {resid.mean():.4f} MPa | resíduos dentro de ±0.8 MPa: "
      f"{np.mean(abs(resid) <= 0.8) * 100:.1f}%")

# ============================================================
# 7. FIGURA 1 (4 painéis)
# ============================================================
r_ens = res_main[res_main.Model == "Ensemble (5 MLPs)"].iloc[0]
fig, ax = plt.subplots(2, 2, figsize=(13, 10))
a = ax[0, 0]
a.scatter(y, ens_pred, s=14, alpha=.6, color="#2f3e4e")
lim = [min(y.min(), ens_pred.min()) - .2, max(y.max(), ens_pred.max()) + .2]
a.plot(lim, lim, "r--", label="1:1")
a.set(xlabel="Observado (MPa)", ylabel="Predito (MPa)",
      title=f"Ensemble (5 MLPs)\nR²={r_ens.R2_pooled:.4f} | RPD={r_ens.RPD_pooled:.2f}")
a.legend()
a = ax[0, 1]
a.scatter(ens_pred, resid, s=14, alpha=.6, color="#e45756")
a.axhline(0, color="r", ls="--")
a.set(xlabel="Predito (MPa)", ylabel="Resíduo (MPa)", title=f"Resíduos | RMSE={r_ens.RMSE_pooled:.4f} MPa")
a = ax[1, 0]
top = res_main.head(7).iloc[::-1]
cols = ["#2f3e4e" if m == "Ensemble (5 MLPs)" else "#5aa9e0" for m in top.Model]
a.barh(top.Model, top.R2_mean, color=cols)
for i, v in enumerate(top.R2_mean):
    a.text(v + .005, i, f"{v:.4f}", va="center", fontsize=8)
a.set(xlabel="R² (média dos folds)", title="Ranking dos modelos", xlim=(0, 1))
a = ax[1, 1]
stages = ["Original (ET, espectro bruto)", "Seleção de atributos (ET)",
          "Voting Ensemble (trees)", "Ensemble (5 MLPs)"]
vals = [res.loc[res.Model == s, "R2_mean"].iloc[0] for s in stages if s in set(res.Model)]
labs = ["Original", "Seleção", "Voting", "MLP Ens."][:len(vals)]
bars = a.bar(labs, vals, color=["#b0bec5", "#5aa9e0", "#f5b650", "#2f3e4e"][:len(vals)])
for b_, v in zip(bars, vals):
    a.text(b_.get_x() + b_.get_width() / 2, v + .01, f"{v:.4f}", ha="center", fontsize=9)
a.set(ylabel="R²", title="Evolução de R² por etapa", ylim=(0, 1))
for lbl, a in zip("abcd", ax.ravel()):
    a.text(-0.08, 1.05, f"({lbl})", transform=a.transAxes, fontsize=13, weight="bold")
plt.tight_layout(); plt.savefig(f"{OUT_DIR}/fig1_resultados.png", dpi=300); plt.show()

# ============================================================
# 8. SALVAR RESULTADOS
# ============================================================
pred_df = pd.DataFrame({"replicate": groups, "y_true": y, "y_pred_ensemble": ens_pred, "residual": resid})
if 'Sistema_Manejo' in df.columns: pred_df["management"] = df['Sistema_Manejo'].values
if 'Genotype' in df.columns: pred_df["genotype"] = df['Genotype'].values

with pd.ExcelWriter(f"{OUT_DIR}/resultados_artigo_043_EAAI.xlsx") as xw:
    res.to_excel(xw, "desempenho_modelos", index=False)
    per_fold.to_excel(xw, "por_fold", index=False)
    err_bins.to_excel(xw, "erro_por_faixa", index=False)
    pred_df.to_excel(xw, "predicoes_oof", index=False)

print(f"\nConcluído. Arquivos salvos em: {OUT_DIR}")
```

```python
# -*- coding: utf-8 -*-
# @title Adiciona os 4 índices espectrais e a coluna Year (ano) aos preditores.
"""
EAAI Pipeline - VERSÃO FINAL
Adiciona os 4 índices espectrais e a coluna Year (ano) aos preditores.
"""

import os, time, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import skew, kurtosis

from sklearn.base import clone
from sklearn.decomposition import PCA
from sklearn.ensemble import ExtraTreesRegressor, RandomForestRegressor, GradientBoostingRegressor
from sklearn.linear_model import Ridge
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from sklearn.cross_decomposition import PLSRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVR

warnings.filterwarnings("ignore")
try:
    from xgboost import XGBRegressor
except Exception: XGBRegressor = None
try:
    from catboost import CatBoostRegressor
except Exception: CatBoostRegressor = None
try:
    from lightgbm import LGBMRegressor
except Exception: LGBMRegressor = None

# from google.colab import drive
# drive.mount("/content/drive")

# ============================================================
# CONFIGURAÇÃO
# ============================================================
BASE_PATH = '/content/drive/MyDrive/Tese/propostas_de_artigos/proposta_artigo_043'
DATA_PATH = os.path.join(BASE_PATH, '/data/dataset_unificado_043.xlsx')
OUT_DIR = os.path.join(BASE_PATH, "/resultado/resultados_artigo_043_FINAL")
os.makedirs(OUT_DIR, exist_ok=True)
np.random.seed(42)
plt.rcParams.update({"figure.dpi": 110, "axes.grid": True, "grid.alpha": .25})

# ============================================================
# 1. LEITURA
# ============================================================
print("Carregando dataset...")
df = pd.read_excel(DATA_PATH)
spec_cols = [c for c in df.columns if str(c).endswith('nm')]
wls = np.array([float(str(c).replace('nm', '')) for c in spec_cols])
order = np.argsort(wls); spec_cols = [spec_cols[i] for i in order]; wls = wls[order]
print(f"Bandas espectrais: {len(spec_cols)} ({wls.min():.1f}-{wls.max():.1f} nm)")

R = df[spec_cols].apply(pd.to_numeric, errors="coerce").interpolate(axis=1, limit_direction="both")
y_all = pd.to_numeric(df['Water_Potential'], errors="coerce")
ok = R.notna().all(axis=1) & y_all.notna() & df['Replicate'].notna()
df, R, y_all = df[ok].reset_index(drop=True), R[ok].reset_index(drop=True), y_all[ok].reset_index(drop=True)
print(f"Amostras válidas: {len(df)}")

y = y_all.values.astype(float)
groups = df['Replicate'].values.astype(int)
raw = R.values.astype(float)

# ============================================================
# 2. ENGENHARIA DE ATRIBUTOS (EAAI + Índices + Year)
# ============================================================
def snv(M):
    return (M - M.mean(1, keepdims=True)) / (M.std(1, keepdims=True) + 1e-12)

X_snv = snv(raw)

# --- ÍNDICES ESPECTRAIS (EAAI Seção 2.4.2) ---
print("Calculando índices espectrais...")
i800, i680 = np.argmin(abs(wls - 800)), np.argmin(abs(wls - 680))
ndvi = (raw[:, i800] - raw[:, i680]) / (raw[:, i800] + raw[:, i680] + 1e-12)
sr = raw[:, i800] / (raw[:, i680] + 1e-12)
idx_feats = np.column_stack([
    ndvi, sr, raw.mean(1), raw.std(1), raw.max(1) - raw.min(1),
    np.percentile(raw, 25, axis=1), np.percentile(raw, 75, axis=1),
    kurtosis(raw, axis=1), skew(raw, axis=1)])
idx_names = ["NDVI", "SR", "mean", "std", "range", "p25", "p75", "kurtosis", "skewness"]

# --- METADADOS (Mês circular + Sistema + ANO) ---
meta_blocks, meta_names = [], []

# Mês circular (2 harmônicos)
mo = df['Month'].values.astype(float)
meta_blocks += [np.sin(2*np.pi*mo/12), np.cos(2*np.pi*mo/12),
                np.sin(4*np.pi*mo/12), np.cos(4*np.pi*mo/12)]
meta_names += ["month_sin1", "month_cos1", "month_sin2", "month_cos2"]

# Sistema de Manejo (Irrigado = 1, Sequeiro = 0)
mgmt = (df['Sistema_Manejo'] == 'Irrigado').astype(float).values
meta_blocks.append(mgmt)
meta_names.append("management")

# ANO (Year) - O EAAI usou o ano como preditor (mesmo o texto dizendo que não)
# No nosso dataset, Year é 14, 15, 16 (que corresponde a 2014, 2015, 2016)
# Vamos usar One-Hot para o ano para capturar efeitos anuais específicos
year_vals = df['Year'].values
for yv in np.unique(year_vals):
    meta_blocks.append((year_vals == yv).astype(float))
    meta_names.append(f"year_{int(yv)}")

meta = np.column_stack(meta_blocks)

# X final = SNV + Índices + Metadados (Mês + Sistema + Ano)
X = np.hstack([X_snv, idx_feats, meta])
names = np.array([f"spec_{w:.1f}" for w in wls] + [f"idx_{n}" for n in idx_names] + [f"meta_{n}" for n in meta_names])
print(f"Matriz de atributos: {X.shape} (bandas={X_snv.shape[1]}, idx={idx_feats.shape[1]}, meta={meta.shape[1]})")

# ============================================================
# 3. PRÉ-PROCESSADOR (Nested dentro do fold)
# ============================================================
class SpectralPreprocessor:
    def __init__(self, n_sel=300, n_pc=30, seed=42):
        self.n_sel, self.n_pc, self.seed = n_sel, n_pc, seed
    def fit(self, X, y):
        self.sc1 = StandardScaler().fit(X)
        Xs = self.sc1.transform(X)
        et = ExtraTreesRegressor(n_estimators=300, max_features=0.33, random_state=self.seed, n_jobs=-1).fit(Xs, y)
        n_sel = min(self.n_sel, X.shape[1])
        self.sel_idx = np.sort(np.argsort(et.feature_importances_)[::-1][:n_sel])
        self.importances_ = et.feature_importances_
        self.pca = PCA(n_components=min(self.n_pc, n_sel), random_state=self.seed)
        Zp = self.pca.fit_transform(Xs[:, self.sel_idx])
        self.sc2 = StandardScaler().fit(Zp)
        return self
    def transform_selected(self, X):
        return self.sc1.transform(X)[:, self.sel_idx]
    def transform(self, X):
        return self.sc2.transform(self.pca.transform(self.transform_selected(X)))

# ============================================================
# 4. MODELOS
# ============================================================
MLP_CONFIGS = [
    ((300, 150, 75, 35), 0.0003, 0.001, 32),
    ((400, 200, 100, 50), 0.0001, 0.001, 64),
    ((300, 150, 75),      0.0005, 0.003, 32),
    ((500, 250, 100),     0.0002, 0.001, 128),
    ((200, 100, 50, 25),  0.0005, 0.005, 32),
]
def make_mlps(seed=42):
    return {f"MLP_{i+1}": MLPRegressor(
        hidden_layer_sizes=a, activation="relu", solver="adam", alpha=al,
        learning_rate_init=lr, batch_size=bs, early_stopping=True,
        validation_fraction=0.15, n_iter_no_change=20, max_iter=2000,
        random_state=seed + i)
        for i, (a, al, lr, bs) in enumerate(MLP_CONFIGS)}

def make_benchmarks(seed=42):
    m = {
        "Extra Trees": ExtraTreesRegressor(n_estimators=300, random_state=seed, n_jobs=-1),
        "Random Forest": RandomForestRegressor(n_estimators=300, random_state=seed, n_jobs=-1),
        "Gradient Boosting": GradientBoostingRegressor(random_state=seed),
        "Ridge": Ridge(alpha=1.0),
        "SVR (RBF)": SVR(C=10),
        "KNN": KNeighborsRegressor(n_neighbors=5),
    }
    if XGBRegressor: m["XGBoost"] = XGBRegressor(random_state=seed, n_jobs=-1, verbosity=0)
    if CatBoostRegressor: m["CatBoost"] = CatBoostRegressor(random_seed=seed, verbose=0)
    if LGBMRegressor: m["LightGBM"] = LGBMRegressor(random_state=seed, verbose=-1)
    return m

def metrics(yt, yp):
    rmse = np.sqrt(mean_squared_error(yt, yp))
    return dict(R2=r2_score(yt, yp), RMSE=rmse, MAE=mean_absolute_error(yt, yp),
                RPD=np.std(yt, ddof=1) / rmse)

def rpd_class(r):
    return "Excelente" if r > 2.5 else "Muito bom" if r >= 2.0 else "Bom" if r >= 1.5 else "Insatisfatório"

# ============================================================
# 5. VALIDAÇÃO LEAVE-ONE-REPLICATE-OUT
# ============================================================
def run_loro(X, raw, names, y, groups):
    n = len(y); uniq = np.unique(groups); store = {}
    def put(model, te, p, t):
        d = store.setdefault(model, dict(pred=np.full(n, np.nan), time={}))
        d["pred"][te] = p; d["time"][groups[te][0]] = t
    for g in uniq:
        te, tr = np.where(groups == g)[0], np.where(groups != g)[0]
        print(f"\n--- Fold: replicata = {g} (treino={len(tr)}, teste={len(te)}) ---")
        prep = SpectralPreprocessor().fit(X[tr], y[tr])
        Ztr, Zte = prep.transform(X[tr]), prep.transform(X[te])

        # Baseline
        t0 = time.time()
        m = ExtraTreesRegressor(n_estimators=300, random_state=42, n_jobs=-1).fit(raw[tr], y[tr])
        put("Original (ET, espectro bruto)", te, m.predict(raw[te]), time.time() - t0)

        # Extra Trees selecionadas
        t0 = time.time()
        Str, Ste = prep.transform_selected(X[tr]), prep.transform_selected(X[te])
        m = ExtraTreesRegressor(n_estimators=300, random_state=42, n_jobs=-1).fit(Str, y[tr])
        put("Seleção de atributos (ET)", te, m.predict(Ste), time.time() - t0)

        # Modelos
        models = make_mlps(); bench = make_benchmarks(); models.update(bench)
        for name, est in models.items():
            t0 = time.time()
            est = clone(est).fit(Ztr, y[tr])
            put(name, te, est.predict(Zte), time.time() - t0)

    # Ensemble
    mlp_names = [f"MLP_{i+1}" for i in range(len(MLP_CONFIGS))]
    ens = np.mean([store[k]["pred"] for k in mlp_names], axis=0)
    ens_time = {g: sum(store[k]["time"][g] for k in mlp_names) for g in uniq}
    store["Ensemble (5 MLPs)"] = dict(pred=ens, time=ens_time)
    # Best MLP
    mean_r2 = {k: np.mean([r2_score(y[groups == g], store[k]["pred"][groups == g]) for g in uniq]) for k in mlp_names}
    best = max(mean_r2, key=mean_r2.get)
    store["MLP (Best Individual)"] = dict(pred=store[best]["pred"].copy(), time=store[best]["time"])
    for k in mlp_names: store[f"[{k}]"] = store.pop(k)
    return store, best

def summarize(store, y, groups):
    rows, per_fold = [], []
    uniq = np.unique(groups)
    for name, d in store.items():
        pf = [metrics(y[groups == g], d["pred"][groups == g]) for g in uniq]
        for g, mm in zip(uniq, pf):
            per_fold.append(dict(Model=name, Fold_test_replicate=g, **mm))
        pooled = metrics(y, d["pred"])
        rows.append(dict(Model=name,
            R2_mean=np.mean([p["R2"] for p in pf]), R2_std=np.std([p["R2"] for p in pf]),
            RMSE_mean=np.mean([p["RMSE"] for p in pf]), MAE_mean=np.mean([p["MAE"] for p in pf]),
            RPD_mean=np.mean([p["RPD"] for p in pf]),
            R2_pooled=pooled["R2"], RMSE_pooled=pooled["RMSE"], RPD_pooled=pooled["RPD"],
            Time_s_per_fold=np.mean(list(d["time"].values()))))
    res = pd.DataFrame(rows).sort_values("R2_mean", ascending=False).reset_index(drop=True)
    res["RPD_class"] = res["RPD_mean"].apply(rpd_class)
    return res, pd.DataFrame(per_fold)

print("\n" + "=" * 70 + "\nVALIDAÇÃO LEAVE-ONE-REPLICATE-OUT\n" + "=" * 70)
store, best_mlp = run_loro(X, raw, names, y, groups)
res, per_fold = summarize(store, y, groups)
res_main = res[~res["Model"].str.startswith("[")]
print(f"\nMelhor MLP individual: {best_mlp}")
print(res_main.round(4).to_string(index=False))

# ============================================================
# 6. ANÁLISE DE RESÍDUOS
# ============================================================
ens_pred = store["Ensemble (5 MLPs)"]["pred"]
resid = y - ens_pred
bins = [(-np.inf, -4), (-4, -2), (-2, -1), (-1, np.inf)]
labels = ["< -4 MPa", "-4 a -2", "-2 a -1", "> -1 MPa"]
rows = []
for (lo, hi), lab in zip(bins, labels):
    m = (y > lo) & (y <= hi)
    if m.sum() > 1:
        rows.append(dict(Faixa=lab, n=int(m.sum()),
                         RMSE=np.sqrt(np.mean(resid[m] ** 2)), MAE=np.mean(abs(resid[m])),
                         Bias=np.mean(resid[m])))
err_bins = pd.DataFrame(rows)
print("\nErro do ensemble por faixa de Ψw:")
print(err_bins.round(3).to_string(index=False))
print(f"Viés global: {resid.mean():.4f} MPa | resíduos dentro de ±0.8 MPa: {np.mean(abs(resid) <= 0.8) * 100:.1f}%")

# ============================================================
# 7. FIGURA 1 (4 painéis)
# ============================================================
r_ens = res_main[res_main.Model == "Ensemble (5 MLPs)"].iloc[0]
fig, ax = plt.subplots(2, 2, figsize=(13, 10))
a = ax[0, 0]
a.scatter(y, ens_pred, s=14, alpha=.6, color="#2f3e4e")
lim = [min(y.min(), ens_pred.min()) - .2, max(y.max(), ens_pred.max()) + .2]
a.plot(lim, lim, "r--", label="1:1")
a.set(xlabel="Observado (MPa)", ylabel="Predito (MPa)",
      title=f"Ensemble (5 MLPs)\nR²={r_ens.R2_pooled:.4f} | RPD={r_ens.RPD_pooled:.2f}")
a.legend()
a = ax[0, 1]
a.scatter(ens_pred, resid, s=14, alpha=.6, color="#e45756")
a.axhline(0, color="r", ls="--")
a.set(xlabel="Predito (MPa)", ylabel="Resíduo (MPa)", title=f"Resíduos | RMSE={r_ens.RMSE_pooled:.4f} MPa")
a = ax[1, 0]
top = res_main.head(7).iloc[::-1]
cols = ["#2f3e4e" if m == "Ensemble (5 MLPs)" else "#5aa9e0" for m in top.Model]
a.barh(top.Model, top.R2_mean, color=cols)
for i, v in enumerate(top.R2_mean):
    a.text(v + .005, i, f"{v:.4f}", va="center", fontsize=8)
a.set(xlabel="R² (média dos folds)", title="Ranking dos modelos", xlim=(0, 1))
a = ax[1, 1]
stages = ["Original (ET, espectro bruto)", "Seleção de atributos (ET)", "Ensemble (5 MLPs)"]
vals = [res.loc[res.Model == s, "R2_mean"].iloc[0] for s in stages if s in set(res.Model)]
labs = ["Original", "Seleção", "MLP Ens."][:len(vals)]
bars = a.bar(labs, vals, color=["#b0bec5", "#5aa9e0", "#2f3e4e"][:len(vals)])
for b_, v in zip(bars, vals):
    a.text(b_.get_x() + b_.get_width() / 2, v + .01, f"{v:.4f}", ha="center", fontsize=9)
a.set(ylabel="R²", title="Evolução de R² por etapa", ylim=(0, 1))
for lbl, a in zip("abcd", ax.ravel()):
    a.text(-0.08, 1.05, f"({lbl})", transform=a.transAxes, fontsize=13, weight="bold")
plt.tight_layout(); plt.savefig(f"{OUT_DIR}/fig1_resultados.png", dpi=300); plt.show()

# ============================================================
# 8. SALVAR RESULTADOS
# ============================================================
pred_df = pd.DataFrame({"replicate": groups, "y_true": y, "y_pred_ensemble": ens_pred, "residual": resid})
if 'Sistema_Manejo' in df.columns: pred_df["management"] = df['Sistema_Manejo'].values
if 'Genotype' in df.columns: pred_df["genotype"] = df['Genotype'].values

with pd.ExcelWriter(f"{OUT_DIR}/resultados_artigo_043_FINAL.xlsx") as xw:
    res.to_excel(xw, "desempenho_modelos", index=False)
    per_fold.to_excel(xw, "por_fold", index=False)
    err_bins.to_excel(xw, "erro_por_faixa", index=False)
    pred_df.to_excel(xw, "predicoes_oof", index=False)

print(f"\nConcluído. Arquivos salvos em: {OUT_DIR}")
```

```python
# -*- coding: utf-8 -*-
#@title EAAI pipeline - V2: experimentos de melhoria + ablação de metadados
"""
EAAI pipeline - V2: experimentos de melhoria + ablação de metadados
Roda várias variantes com Leave-One-Replicate-Out, compara numa tabela e
faz uma rodada final (bagging de MLPs) com a melhor variante (exceto ablações).

Atenção: escolher a melhor variante pelo R² dos próprios folds é levemente
otimista. Relate também as linhas de ablação (X1-X3).
"""
import os, time, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import skew, kurtosis
from scipy.signal import savgol_filter
from sklearn.decomposition import PCA
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler

warnings.filterwarnings("ignore")
np.random.seed(42)
plt.rcParams.update({"figure.dpi": 110, "axes.grid": True, "grid.alpha": .25})

# ============================================================
# CONFIG
# ============================================================
BASE_PATH = '/content/drive/MyDrive/Tese/propostas_de_artigos/proposta_artigo_043'
DATA_PATH = os.path.join(BASE_PATH, '/data/dataset_unificado_043.xlsx')
OUT_DIR = os.path.join(BASE_PATH, "/resultado/resultados_artigo_043_V2")
os.makedirs(OUT_DIR, exist_ok=True)

DROP_POSITIVE_PSI = False   # True remove Ψw > 0 (fisicamente suspeito)
N_SEEDS_EXPERIMENTS = 1     # sementes por arquitetura nos experimentos (rápido)
N_SEEDS_FINAL = 5           # sementes por arquitetura na rodada final (bagging)

# ============================================================
# 1. LEITURA
# ============================================================
df = pd.read_excel(DATA_PATH)
spec_cols = [c for c in df.columns if str(c).endswith('nm')]
wls_all = np.array([float(str(c).replace('nm', '')) for c in spec_cols])
o = np.argsort(wls_all); spec_cols = [spec_cols[i] for i in o]; wls_all = wls_all[o]

R = df[spec_cols].apply(pd.to_numeric, errors="coerce").interpolate(axis=1, limit_direction="both")
y_all = pd.to_numeric(df['Water_Potential'], errors="coerce")
ok = R.notna().all(axis=1) & y_all.notna() & df['Replicate'].notna()
if DROP_POSITIVE_PSI:
    ok &= (y_all <= 0)
df, R, y_all = df[ok].reset_index(drop=True), R[ok].reset_index(drop=True), y_all[ok].reset_index(drop=True)

y = y_all.values.astype(float)
groups = df['Replicate'].values.astype(int)
raw_all = R.values.astype(float)
uniq = np.unique(groups)
n = len(y)
print(f"Amostras: {n} | bandas: {len(wls_all)} ({wls_all.min():.0f}-{wls_all.max():.0f} nm)")
print(f"Diagnóstico: Ψw>0: {(y > 0).sum()} | Ψw<=-6.5: {(y <= -6.5).sum()} "
      f"(valores únicos nessa faixa: {np.unique(np.round(y[y <= -6.5], 2))[:6]})")

# ============================================================
# 2. FEATURES
# ============================================================
def snv(M):
    return (M - M.mean(1, keepdims=True)) / (M.std(1, keepdims=True) + 1e-12)

def bin_spectra(raw, wls, wl_max, bin_nm):
    m = wls <= wl_max
    r, w = raw[:, m], wls[m]
    if bin_nm and bin_nm > 0:
        edges = np.arange(w.min(), w.max() + bin_nm, bin_nm)
        idx = np.digitize(w, edges) - 1
        u = np.unique(idx)
        r = np.column_stack([r[:, idx == k].mean(1) for k in u])
        w = np.array([w[idx == k].mean() for k in u])
    return r, w

_cache = {}
def build_features(wl_max=900, bin_nm=0, deriv=False, extra_idx=False, meta='month_mgmt_year'):
    key = (wl_max, bin_nm, deriv, extra_idx, meta)
    if key in _cache:
        return _cache[key]
    r, w = bin_spectra(raw_all, wls_all, wl_max, bin_nm)
    near = lambda nm: int(np.argmin(abs(w - nm)))
    Rn = lambda nm: r[:, near(nm)]

    blocks = [snv(r)]
    names = [f"spec_{x:.1f}" for x in w]

    # índices base (artigo)
    idx = {"NDVI": (Rn(800) - Rn(680)) / (Rn(800) + Rn(680) + 1e-12),
           "SR": Rn(800) / (Rn(680) + 1e-12),
           "mean": r.mean(1), "std": r.std(1), "range": r.max(1) - r.min(1),
           "p25": np.percentile(r, 25, axis=1), "p75": np.percentile(r, 75, axis=1),
           "kurtosis": kurtosis(r, axis=1), "skewness": skew(r, axis=1)}
    if extra_idx:
        d = np.gradient(r, w, axis=1)
        mk = (w >= 680) & (w <= 750)
        idx["REP"] = w[mk][np.argmax(d[:, mk], axis=1)]
        idx["NDRE"] = (Rn(750) - Rn(710)) / (Rn(750) + Rn(710) + 1e-12)
        idx["CIre"] = Rn(750) / (Rn(710) + 1e-12) - 1
        idx["mND705"] = (Rn(750) - Rn(705)) / (Rn(750) + Rn(705) - 2 * Rn(445) + 1e-12)
        idx["PSRI"] = (Rn(680) - Rn(500)) / (Rn(750) + 1e-12)
        idx["PRI"] = (Rn(531) - Rn(570)) / (Rn(531) + Rn(570) + 1e-12)
        if w.max() >= 945:
            idx["WBI"] = Rn(950) / (Rn(900) + 1e-12)
    blocks.append(np.column_stack(list(idx.values())))
    names += [f"idx_{k}" for k in idx]

    if deriv:
        spacing = np.median(np.diff(w))
        win = max(5, (int(round(15 / spacing)) // 2) * 2 + 1)
        win = min(win, (r.shape[1] // 2) * 2 - 1)
        d1 = savgol_filter(r, window_length=win, polyorder=2, deriv=1, delta=spacing, axis=1)
        blocks.append(d1)
        names += [f"d1_{x:.1f}" for x in w]

    mo = df['Month'].values.astype(float)
    mg = (df['Sistema_Manejo'] == 'Irrigado').astype(float).values
    mb = []
    if meta in ('month_mgmt', 'month_mgmt_year'):
        mb += [np.sin(2*np.pi*mo/12), np.cos(2*np.pi*mo/12), np.sin(4*np.pi*mo/12), np.cos(4*np.pi*mo/12)]
        names += ["meta_month_sin1", "meta_month_cos1", "meta_month_sin2", "meta_month_cos2"]
    if meta in ('mgmt', 'month_mgmt', 'month_mgmt_year'):
        mb.append(mg); names.append("meta_management")
    if meta == 'month_mgmt_year':
        for yv in np.unique(df['Year'].values):
            mb.append((df['Year'].values == yv).astype(float)); names.append(f"meta_year_{int(yv)}")
    if mb:
        blocks.append(np.column_stack(mb))
    X = np.hstack(blocks)
    _cache[key] = (X, np.array(names))
    return _cache[key]

# ============================================================
# 3. PRÉ-PROCESSADOR E MODELOS
# ============================================================
class Prep:
    def __init__(self, n_sel=300, n_pc=30, seed=42):
        self.n_sel, self.n_pc, self.seed = n_sel, n_pc, seed
    def fit(self, X, y):
        self.sc1 = StandardScaler().fit(X)
        Xs = self.sc1.transform(X)
        et = ExtraTreesRegressor(n_estimators=300, max_features=0.33,
                                 random_state=self.seed, n_jobs=-1).fit(Xs, y)
        k = min(self.n_sel, X.shape[1])
        self.sel = np.sort(np.argsort(et.feature_importances_)[::-1][:k])
        if self.n_pc:
            self.pca = PCA(min(self.n_pc, k), random_state=self.seed).fit(Xs[:, self.sel])
            self.sc2 = StandardScaler().fit(self.pca.transform(Xs[:, self.sel]))
        return self
    def sel_tf(self, X):
        return self.sc1.transform(X)[:, self.sel]
    def tf(self, X):
        Z = self.sel_tf(X)
        return Z if not self.n_pc else self.sc2.transform(self.pca.transform(Z))

MLP_CONFIGS = [((300, 150, 75, 35), 0.0003, 0.001, 32), ((400, 200, 100, 50), 0.0001, 0.001, 64),
               ((300, 150, 75), 0.0005, 0.003, 32), ((500, 250, 100), 0.0002, 0.001, 128),
               ((200, 100, 50, 25), 0.0005, 0.005, 32)]

fwd = lambda v: np.log1p(-v)        # Ψ (≤0.2) -> log(1-Ψ)
inv = lambda z: 1.0 - np.exp(z)

def fit_predict_fold(Xtr, ytr, Xte, prep_kw, log_target, n_seeds, et_leaf):
    prep = Prep(**prep_kw).fit(Xtr, ytr)
    Str, Ste = prep.sel_tf(Xtr), prep.sel_tf(Xte)
    Ztr, Zte = prep.tf(Xtr), prep.tf(Xte)
    yt = fwd(ytr) if log_target else ytr
    et = ExtraTreesRegressor(n_estimators=500, min_samples_leaf=et_leaf,
                             random_state=42, n_jobs=-1).fit(Str, yt)
    p_et = et.predict(Ste)
    preds = []
    for s in range(n_seeds):
        for i, (a, al, lr, bs) in enumerate(MLP_CONFIGS):
            m = MLPRegressor(hidden_layer_sizes=a, activation="relu", solver="adam", alpha=al,
                             learning_rate_init=lr, batch_size=bs, early_stopping=True,
                             validation_fraction=0.15, n_iter_no_change=20, max_iter=2000,
                             random_state=42 + i + 100 * s).fit(Ztr, yt)
            preds.append(m.predict(Zte))
    p_mlp = np.mean(preds, axis=0)
    if log_target:
        p_et, p_mlp = inv(p_et), inv(p_mlp)
    return p_et, p_mlp

def met(yt, yp):
    rmse = np.sqrt(mean_squared_error(yt, yp))
    return dict(R2=r2_score(yt, yp), RMSE=rmse, MAE=mean_absolute_error(yt, yp),
                RPD=np.std(yt, ddof=1) / rmse)

def evaluate(label, feat_kw, prep_kw, log_target=False, n_seeds=1, et_leaf=1):
    X, names = build_features(**feat_kw)
    t0 = time.time()
    P = {k: np.zeros(n) for k in ("ET (sel.)", "MLP Ens.", "Blend ET+MLP")}
    for g in uniq:
        te, tr = np.where(groups == g)[0], np.where(groups != g)[0]
        pe, pm = fit_predict_fold(X[tr], y[tr], X[te], prep_kw, log_target, n_seeds, et_leaf)
        P["ET (sel.)"][te], P["MLP Ens."][te], P["Blend ET+MLP"][te] = pe, pm, (pe + pm) / 2
    row = dict(Variante=label, n_feat=X.shape[1])
    for k, p in P.items():
        fold_r2 = [r2_score(y[groups == g], p[groups == g]) for g in uniq]
        row[f"R2_{k}"] = np.mean(fold_r2)
        if k == "Blend ET+MLP":
            row["R2_std"] = np.std(fold_r2)
            row["R2_Rep4"] = fold_r2[list(uniq).index(4)] if 4 in uniq else np.nan
            pm_ = met(y, p); row["RMSE"] = pm_["RMSE"]; row["RPD"] = pm_["RPD"]
    row["tempo_s"] = time.time() - t0
    print(f"  {label:48s} ET={row['R2_ET (sel.)']:.4f}  MLP={row['R2_MLP Ens.']:.4f}  "
          f"Blend={row['R2_Blend ET+MLP']:.4f}  ({row['tempo_s']:.0f}s)")
    return row, P

# ============================================================
# 4. EXPERIMENTOS
# ============================================================
base = dict(wl_max=951, bin_nm=0, deriv=False, extra_idx=False, meta='month_mgmt_year')
pk = dict(n_sel=300, n_pc=30)
D = {**base, 'wl_max': 900, 'bin_nm': 5, 'deriv': True, 'extra_idx': True}

VARIANTS = [
    ("A  baseline (seu script, 951 nm)",         base,                                pk,                         False),
    ("B  só até 900 nm",                         {**base, 'wl_max': 900},             pk,                         False),
    ("C  900 nm + bin 5 nm",                     {**base, 'wl_max': 900, 'bin_nm': 5}, pk,                        False),
    ("D  C + derivada SG + índices extras",      D,                                   pk,                         False),
    ("D2 951 nm + bin 5 + deriv + extras (WBI)", {**D, 'wl_max': 951},                pk,                         False),
    ("E  D + alvo log(1-Ψ)",                     D,                                   pk,                         True),
    ("F  D, sem PCA (MLP nas 300 sel.)",         D,                                   dict(n_sel=300, n_pc=None), False),
    ("G  D, n_sel=100, n_pc=20",                 D,                                   dict(n_sel=100, n_pc=20),   False),
    ("H  D, n_sel=200, n_pc=40",                 D,                                   dict(n_sel=200, n_pc=40),   False),
    # --- ablação de metadados (resultado MAIS importante para o artigo) ---
    ("X1 D sem metadados (só espectro)",         {**D, 'meta': 'none'},               pk,                         False),
    ("X2 D só manejo",                           {**D, 'meta': 'mgmt'},               pk,                         False),
    ("X3 D mês+manejo (sem ano)",                {**D, 'meta': 'month_mgmt'},         pk,                         False),
]

print("\n" + "=" * 90 + "\nEXPERIMENTOS (R² = média dos 4 folds LORO)\n" + "=" * 90)
rows, preds_store = [], {}
for lab, fk, pkw, lg in VARIANTS:
    r_, P_ = evaluate(lab, fk, pkw, log_target=lg, n_seeds=N_SEEDS_EXPERIMENTS)
    rows.append(r_); preds_store[lab] = P_
exp = pd.DataFrame(rows)
print("\n" + exp.round(4).to_string(index=False))
exp.to_csv(f"{OUT_DIR}/experimentos.csv", index=False)

# ============================================================
# 5. RODADA FINAL (bagging de MLPs) na melhor variante (exceto ablações)
# ============================================================
cand = exp[~exp.Variante.str.startswith("X")]
best_label = cand.loc[cand["R2_Blend ET+MLP"].idxmax(), "Variante"]
fk, pkw, lg = [(v[1], v[2], v[3]) for v in VARIANTS if v[0] == best_label][0]
print(f"\n>>> Variante escolhida: {best_label}  (bagging: {N_SEEDS_FINAL} sementes x 5 arquiteturas)")
row_f, P_f = evaluate("FINAL " + best_label, fk, pkw, log_target=lg, n_seeds=N_SEEDS_FINAL)

pf = []
for k, p in P_f.items():
    for g in uniq:
        pf.append(dict(Modelo=k, Replicata_teste=g, **met(y[groups == g], p[groups == g])))
pf = pd.DataFrame(pf)
print("\nDesempenho por fold (compare a replicata 4 com o artigo):")
print(pf.round(4).to_string(index=False))

pred = P_f["Blend ET+MLP"]
resid = y - pred
bins = [(-np.inf, -4), (-4, -2), (-2, -1), (-1, np.inf)]
labs = ["< -4 MPa", "-4 a -2", "-2 a -1", "> -1 MPa"]
eb = []
for (lo, hi), lb in zip(bins, labs):
    m = (y > lo) & (y <= hi)
    if m.sum() > 1:
        eb.append(dict(Faixa=lb, n=int(m.sum()), RMSE=np.sqrt(np.mean(resid[m] ** 2)),
                       MAE=np.mean(abs(resid[m])), Bias=np.mean(resid[m]),
                       pct_SSE=100 * np.sum(resid[m] ** 2) / np.sum(resid ** 2)))
eb = pd.DataFrame(eb)
print("\nErro por faixa de Ψw (pct_SSE = % do erro quadrático total):")
print(eb.round(3).to_string(index=False))

# Figura
fig, ax = plt.subplots(2, 2, figsize=(13, 10))
a = ax[0, 0]; a.scatter(y, pred, s=14, alpha=.6, color="#2f3e4e")
lim = [min(y.min(), pred.min()) - .2, max(y.max(), pred.max()) + .2]; a.plot(lim, lim, "r--")
mm = met(y, pred)
a.set(xlabel="Observado (MPa)", ylabel="Predito (MPa)",
      title=f"Blend ET+MLP | R²={mm['R2']:.4f} | RPD={mm['RPD']:.2f}")
a = ax[0, 1]; a.scatter(pred, resid, s=14, alpha=.6, color="#e45756"); a.axhline(0, color="r", ls="--")
a.set(xlabel="Predito (MPa)", ylabel="Resíduo (MPa)", title=f"Resíduos | RMSE={mm['RMSE']:.4f} MPa")
a = ax[1, 0]
e2 = exp.sort_values("R2_Blend ET+MLP")
a.barh(e2.Variante.str.slice(0, 34), e2["R2_Blend ET+MLP"], color="#5aa9e0")
a.set(xlabel="R² (média dos folds) - Blend", title="Variantes (X = ablação de metadados)", xlim=(0, 1))
a.tick_params(axis='y', labelsize=7)
a = ax[1, 1]; a.bar(eb.Faixa, eb.RMSE, color="#e45756"); a.set(ylabel="RMSE (MPa)", title="Erro por faixa de Ψw")
for lbl, a in zip("abcd", ax.ravel()):
    a.text(-0.08, 1.05, f"({lbl})", transform=a.transAxes, fontsize=13, weight="bold")
plt.tight_layout(); plt.savefig(f"{OUT_DIR}/fig_v2.png", dpi=300); plt.show()

with pd.ExcelWriter(f"{OUT_DIR}/resultados_v2.xlsx") as xw:
    exp.to_excel(xw, "experimentos", index=False)
    pf.to_excel(xw, "final_por_fold", index=False)
    eb.to_excel(xw, "final_erro_por_faixa", index=False)
    pd.DataFrame({"replicate": groups, "y_true": y, "y_pred_blend": pred, "y_pred_ET": P_f["ET (sel.)"],
                  "y_pred_MLP": P_f["MLP Ens."], "residual": resid}).to_excel(xw, "predicoes_oof", index=False)
print(f"\nConcluído. Arquivos em: {OUT_DIR}")
```

```python
# -*- coding: utf-8 -*-
#@title EAAI pipeline - V3
"""
EAAI pipeline - V3
 - Configuração C (406-900 nm, janelas de 5 nm), SEM o ano como preditor
 - Cenários: 'month_mgmt' (mês + manejo) e 'none' (só espectro)
 - Modelos: ET (selecionado), ensemble de 5 MLPs, SVR; blends
 - Calibração linear ANINHADA (ajustada só com predições internas do treino)
 - Intervalos de confiança por bootstrap em blocos (réplica x ano x mês)
 - Desempenho dentro de cada manejo
Não testado: rode e ajuste se necessário.
"""
import os, time, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import skew, kurtosis
from sklearn.decomposition import PCA
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVR

warnings.filterwarnings("ignore")
np.random.seed(42)
plt.rcParams.update({"figure.dpi": 110, "axes.grid": True, "grid.alpha": .25})

# ============================================================
# CONFIG
# ============================================================
BASE_PATH = '/content/drive/MyDrive/Tese/propostas_de_artigos/proposta_artigo_043'
DATA_PATH = os.path.join(BASE_PATH, '/data/dataset_unificado_043.xlsx')
OUT_DIR = os.path.join(BASE_PATH, "/resultado/resultados_artigo_043_V3")
os.makedirs(OUT_DIR, exist_ok=True)

WL_MIN, WL_MAX, BIN_NM = 406.0, 900.0, 5
SCENARIOS = ['month_mgmt', 'none']
N_SEEDS = 3          # sementes por arquitetura no ajuste externo
N_BOOT = 1000
EXCLUDE_EXTREMES = False   # True: métricas também sem Ψw <= -6.5 e > 0 (reportadas à parte)

# ============================================================
# 1. LEITURA
# ============================================================
df = pd.read_excel(DATA_PATH)
spec_cols = [c for c in df.columns if str(c).endswith('nm')]
wls_all = np.array([float(str(c).replace('nm', '')) for c in spec_cols])
o = np.argsort(wls_all); spec_cols = [spec_cols[i] for i in o]; wls_all = wls_all[o]
R = df[spec_cols].apply(pd.to_numeric, errors="coerce").interpolate(axis=1, limit_direction="both")
y_all = pd.to_numeric(df['Water_Potential'], errors="coerce")
ok = R.notna().all(axis=1) & y_all.notna() & df['Replicate'].notna()
df, R, y_all = df[ok].reset_index(drop=True), R[ok].reset_index(drop=True), y_all[ok].reset_index(drop=True)
y = y_all.values.astype(float)
groups = df['Replicate'].values.astype(int)
raw_all = R.values.astype(float)
uniq, n = np.unique(groups), len(y)
print(f"Amostras: {n} | replicatas: {list(uniq)}")

# ============================================================
# 2. FEATURES
# ============================================================
def snv(M):
    return (M - M.mean(1, keepdims=True)) / (M.std(1, keepdims=True) + 1e-12)

def build_features(meta):
    m = (wls_all >= WL_MIN) & (wls_all <= WL_MAX)
    r, w = raw_all[:, m], wls_all[m]
    edges = np.arange(w.min(), w.max() + BIN_NM, BIN_NM)
    idx = np.digitize(w, edges) - 1
    u = np.unique(idx)
    r = np.column_stack([r[:, idx == k].mean(1) for k in u])
    w = np.array([w[idx == k].mean() for k in u])
    near = lambda nm: int(np.argmin(abs(w - nm)))
    ndvi = (r[:, near(800)] - r[:, near(680)]) / (r[:, near(800)] + r[:, near(680)] + 1e-12)
    sr = r[:, near(800)] / (r[:, near(680)] + 1e-12)
    ind = np.column_stack([ndvi, sr, r.mean(1), r.std(1), r.max(1) - r.min(1),
                           np.percentile(r, 25, axis=1), np.percentile(r, 75, axis=1),
                           kurtosis(r, axis=1), skew(r, axis=1)])
    blocks = [snv(r), ind]
    mo = df['Month'].values.astype(float)
    mg = (df['Sistema_Manejo'] == 'Irrigado').astype(float).values
    if meta == 'month_mgmt':
        blocks.append(np.column_stack([np.sin(2*np.pi*mo/12), np.cos(2*np.pi*mo/12),
                                       np.sin(4*np.pi*mo/12), np.cos(4*np.pi*mo/12), mg]))
    return np.hstack(blocks)

# ============================================================
# 3. MODELOS
# ============================================================
class Prep:
    def __init__(self, n_sel=300, n_pc=30, seed=42):
        self.n_sel, self.n_pc, self.seed = n_sel, n_pc, seed
    def fit(self, X, y):
        self.sc1 = StandardScaler().fit(X); Xs = self.sc1.transform(X)
        et = ExtraTreesRegressor(300, max_features=0.33, random_state=self.seed, n_jobs=-1).fit(Xs, y)
        k = min(self.n_sel, X.shape[1])
        self.sel = np.sort(np.argsort(et.feature_importances_)[::-1][:k])
        self.pca = PCA(min(self.n_pc, k), random_state=self.seed).fit(Xs[:, self.sel])
        self.sc2 = StandardScaler().fit(self.pca.transform(Xs[:, self.sel]))
        return self
    def sel_tf(self, X): return self.sc1.transform(X)[:, self.sel]
    def tf(self, X): return self.sc2.transform(self.pca.transform(self.sel_tf(X)))

MLP_CONFIGS = [((300, 150, 75, 35), 0.0003, 0.001, 32), ((400, 200, 100, 50), 0.0001, 0.001, 64),
               ((300, 150, 75), 0.0005, 0.003, 32), ((500, 250, 100), 0.0002, 0.001, 128),
               ((200, 100, 50, 25), 0.0005, 0.005, 32)]
COMBOS = ['ET', 'MLP', 'SVR', 'ET+MLP', 'ET+MLP+SVR']

def fit_predict(Xtr, ytr, Xte, n_seeds):
    prep = Prep().fit(Xtr, ytr)
    Str, Ste, Ztr, Zte = prep.sel_tf(Xtr), prep.sel_tf(Xte), prep.tf(Xtr), prep.tf(Xte)
    p_et = ExtraTreesRegressor(500, random_state=42, n_jobs=-1).fit(Str, ytr).predict(Ste)
    ml = []
    for s in range(n_seeds):
        for i, (a, al, lr, bs) in enumerate(MLP_CONFIGS):
            mdl = MLPRegressor(hidden_layer_sizes=a, activation="relu", solver="adam", alpha=al,
                               learning_rate_init=lr, batch_size=bs, early_stopping=True,
                               validation_fraction=0.15, n_iter_no_change=20, max_iter=2000,
                               random_state=42 + i + 100 * s).fit(Ztr, ytr)
            ml.append(mdl.predict(Zte))
    p_mlp = np.mean(ml, axis=0)
    p_svr = SVR(C=10).fit(Ztr, ytr).predict(Zte)
    return {'ET': p_et, 'MLP': p_mlp, 'SVR': p_svr,
            'ET+MLP': (p_et + p_mlp) / 2, 'ET+MLP+SVR': (p_et + p_mlp + p_svr) / 3}

def met(yt, yp):
    rmse = np.sqrt(mean_squared_error(yt, yp))
    return dict(R2=r2_score(yt, yp), RMSE=rmse, RPD=np.std(yt, ddof=1) / rmse)

# ============================================================
# 4. VALIDAÇÃO LORO COM CALIBRAÇÃO ANINHADA
# ============================================================
def run_scenario(meta):
    X = build_features(meta)
    raw_p = {k: np.zeros(n) for k in COMBOS}
    cal_p = {k: np.zeros(n) for k in COMBOS}
    slopes = []
    t0 = time.time()
    for g in uniq:
        te, tr = np.where(groups == g)[0], np.where(groups != g)[0]
        P = fit_predict(X[tr], y[tr], X[te], N_SEEDS)
        inner = {k: np.zeros(len(tr)) for k in COMBOS}       # OOF interno só com o treino
        for h in np.unique(groups[tr]):
            ite = np.where(groups[tr] == h)[0]; itr = np.where(groups[tr] != h)[0]
            Q = fit_predict(X[tr][itr], y[tr][itr], X[tr][ite], 1)
            for k in COMBOS:
                inner[k][ite] = Q[k]
        for k in COMBOS:
            lr = LinearRegression().fit(inner[k].reshape(-1, 1), y[tr])
            raw_p[k][te] = P[k]
            cal_p[k][te] = lr.predict(P[k].reshape(-1, 1))
            slopes.append(dict(Fold=g, Modelo=k, intercepto=lr.intercept_, inclinacao=lr.coef_[0]))
        print(f"  [{meta}] fold {g} ok ({time.time() - t0:.0f}s)")
    return raw_p, cal_p, pd.DataFrame(slopes)

def table(preds, label):
    rows = []
    for k, p in preds.items():
        fr = [r2_score(y[groups == g], p[groups == g]) for g in uniq]
        m = met(y, p); tail = y <= -4
        rows.append({"Modelo": k, "Versao": label, "R2_mean": np.mean(fr), "R2_std": np.std(fr),
                     "R2_Rep4": fr[list(uniq).index(4)] if 4 in uniq else np.nan,
                     "RMSE": m["RMSE"], "RPD": m["RPD"],
                     "RMSE_<-4": np.sqrt(np.mean((y[tail] - p[tail]) ** 2)),
                     "Vies_<-4": np.mean(y[tail] - p[tail])})
    return pd.DataFrame(rows)

# bootstrap em blocos (réplica x ano x mês)
cl = pd.factorize(pd.Series(list(zip(groups, df['Year'], df['Month']))))[0]
cl_idx = [np.where(cl == c)[0] for c in range(cl.max() + 1)]

def boot(pred_a, pred_b=None, B=N_BOOT, seed=0):
    rng = np.random.default_rng(seed); ra, dd = [], []
    for _ in range(B):
        pick = rng.integers(0, len(cl_idx), len(cl_idx))
        ix = np.concatenate([cl_idx[i] for i in pick])
        a = r2_score(y[ix], pred_a[ix]); ra.append(a)
        if pred_b is not None:
            dd.append(a - r2_score(y[ix], pred_b[ix]))
    ci = np.percentile(ra, [2.5, 97.5])
    di = np.percentile(dd, [2.5, 97.5]) if dd else (np.nan, np.nan)
    return ci, di

results, all_tabs, all_slopes = {}, [], []
for sc in SCENARIOS:
    print("\n" + "=" * 80 + f"\nCENÁRIO: {sc}\n" + "=" * 80)
    raw_p, cal_p, sl = run_scenario(sc)
    results[sc] = (raw_p, cal_p)
    tab = pd.concat([table(raw_p, "bruto"), table(cal_p, "calibrado")]).assign(Cenario=sc)
    all_tabs.append(tab); all_slopes.append(sl.assign(Cenario=sc))
    print("\n" + tab.round(4).to_string(index=False))
    print("\nCalibração (inclinação média por modelo; >1 = modelo encolhia para a média):")
    print(sl.groupby("Modelo")[["intercepto", "inclinacao"]].mean().round(3).to_string())

    ref = raw_p['ET+MLP']
    print("\nIC 95% (bootstrap em blocos) do R² e da diferença contra 'ET+MLP' bruto:")
    for lab, p in [("ET+MLP bruto", raw_p['ET+MLP']), ("ET+MLP calibrado", cal_p['ET+MLP']),
                   ("ET+MLP+SVR bruto", raw_p['ET+MLP+SVR']), ("ET+MLP+SVR calibrado", cal_p['ET+MLP+SVR']),
                   ("MLP bruto", raw_p['MLP'])]:
        ci, di = boot(p, None if lab == "ET+MLP bruto" else ref)
        print(f"  {lab:24s} R2 IC95%=[{ci[0]:.3f}, {ci[1]:.3f}]"
              + ("" if lab == "ET+MLP bruto" else f" | diferença IC95%=[{di[0]:+.3f}, {di[1]:+.3f}]"))

    best_p = cal_p['ET+MLP+SVR']
    print("\nDesempenho dentro de cada manejo (ET+MLP+SVR calibrado):")
    for v in df['Sistema_Manejo'].unique():
        m = (df['Sistema_Manejo'] == v).values
        print(f"  {v:10s} n={m.sum():4d} R2={r2_score(y[m], best_p[m]):.3f} "
              f"RMSE={np.sqrt(np.mean((y[m] - best_p[m]) ** 2)):.3f} SD(y)={y[m].std(ddof=1):.3f}")

    if EXCLUDE_EXTREMES:
        keep = (y > -6.5) & (y <= 0)
        print(f"\nSem extremos (n={keep.sum()}): R2={r2_score(y[keep], best_p[keep]):.3f} "
              f"RMSE={np.sqrt(np.mean((y[keep] - best_p[keep]) ** 2)):.3f}")

# ============================================================
# 5. FIGURA: antes x depois da calibração (cenário principal)
# ============================================================
sc0 = SCENARIOS[0]
raw_p, cal_p = results[sc0]
fig, ax = plt.subplots(1, 2, figsize=(12, 5))
for a, (lab, p) in zip(ax, [("bruto", raw_p['ET+MLP+SVR']), ("calibrado", cal_p['ET+MLP+SVR'])]):
    a.scatter(y, p, s=14, alpha=.6, color="#2f3e4e")
    lim = [min(y.min(), p.min()) - .2, max(y.max(), p.max()) + .2]
    a.plot(lim, lim, "r--")
    mm = met(y, p)
    a.set(xlabel="Observado (MPa)", ylabel="Predito (MPa)",
          title=f"ET+MLP+SVR {lab} [{sc0}]\nR²={mm['R2']:.4f} | RMSE={mm['RMSE']:.3f} | RPD={mm['RPD']:.2f}")
plt.tight_layout(); plt.savefig(f"{OUT_DIR}/fig_v3_calibracao.png", dpi=300); plt.show()

with pd.ExcelWriter(f"{OUT_DIR}/resultados_v3.xlsx") as xw:
    pd.concat(all_tabs).to_excel(xw, "desempenho", index=False)
    pd.concat(all_slopes).to_excel(xw, "calibracao", index=False)
    pd.DataFrame({"replicate": groups, "y_true": y,
                  **{f"{k}_bruto": raw_p[k] for k in COMBOS},
                  **{f"{k}_calibrado": cal_p[k] for k in COMBOS}}).to_excel(xw, f"preds_{sc0}", index=False)
print(f"\nConcluído. Arquivos em: {OUT_DIR}")
```

```python
# -*- coding: utf-8 -*-
# @title EAAI pipeline - V4: quanto o ESPECTRO acrescenta além de mês + manejo?
"""
EAAI pipeline - V4: quanto o ESPECTRO acrescenta além de mês + manejo?

Modelos comparados (todos com validação Leave-One-Replicate-Out):
  1. Espectro + metadados (mês + manejo)   : ET, ensemble de 5 MLPs, ET+MLP
  2. Só espectro                           : ET, ensemble de 5 MLPs, ET+MLP
  3. Só metadados (mês + manejo)           : Extra Trees (min_samples_leaf=5)
  4. Só metadados, tabela de médias        : média de Ψw por (mês x manejo) do treino

Saídas:
  - R² por fold, global e dentro de cada manejo (Irrigado / Sequeiro)
  - IC 95% por bootstrap em blocos (réplica x ano x mês) do R² e das DIFERENÇAS
    entre modelos (global e dentro de cada manejo)
  - figura e Excel em OUT_DIR
Configuração espectral: 406-900 nm, janelas de 5 nm, SEM o ano como preditor.
Não testado: rode e ajuste se necessário.
"""
import os, time, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import skew, kurtosis
from sklearn.decomposition import PCA
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.metrics import r2_score, mean_squared_error
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler

warnings.filterwarnings("ignore")
np.random.seed(42)
plt.rcParams.update({"figure.dpi": 110, "axes.grid": True, "grid.alpha": .25})

# ============================================================
# CONFIG
# ============================================================
BASE_PATH = '/content/drive/MyDrive/Tese/propostas_de_artigos/proposta_artigo_043'
DATA_PATH = os.path.join(BASE_PATH, '/data/dataset_unificado_043.xlsx')
OUT_DIR = os.path.join(BASE_PATH, "/resultado/resultados_artigo_043_V4")
os.makedirs(OUT_DIR, exist_ok=True)

WL_MIN, WL_MAX, BIN_NM = 406.0, 900.0, 5
N_SEEDS = 3
N_BOOT = 1000

# ============================================================
# 1. LEITURA
# ============================================================
df = pd.read_excel(DATA_PATH)
spec_cols = [c for c in df.columns if str(c).endswith('nm')]
wls_all = np.array([float(str(c).replace('nm', '')) for c in spec_cols])
o = np.argsort(wls_all); spec_cols = [spec_cols[i] for i in o]; wls_all = wls_all[o]
R = df[spec_cols].apply(pd.to_numeric, errors="coerce").interpolate(axis=1, limit_direction="both")
y_all = pd.to_numeric(df['Water_Potential'], errors="coerce")
ok = R.notna().all(axis=1) & y_all.notna() & df['Replicate'].notna()
df, R, y_all = df[ok].reset_index(drop=True), R[ok].reset_index(drop=True), y_all[ok].reset_index(drop=True)
y = y_all.values.astype(float)
groups = df['Replicate'].values.astype(int)
raw_all = R.values.astype(float)
uniq, n = np.unique(groups), len(y)

mo = df['Month'].values.astype(float)
mg = (df['Sistema_Manejo'] == 'Irrigado').astype(float).values
MG = mg == 1
masks = {"Global": np.ones(n, bool), "Irrigado": MG, "Sequeiro": ~MG}
print(f"Amostras: {n} | irrigado={MG.sum()} | sequeiro={(~MG).sum()} | replicatas={list(uniq)}")

# ============================================================
# 2. FEATURES ESPECTRAIS
# ============================================================
def snv(M):
    return (M - M.mean(1, keepdims=True)) / (M.std(1, keepdims=True) + 1e-12)

META_BLOCK = np.column_stack([np.sin(2*np.pi*mo/12), np.cos(2*np.pi*mo/12),
                              np.sin(4*np.pi*mo/12), np.cos(4*np.pi*mo/12), mg])

def build_features(with_meta):
    m = (wls_all >= WL_MIN) & (wls_all <= WL_MAX)
    r, w = raw_all[:, m], wls_all[m]
    edges = np.arange(w.min(), w.max() + BIN_NM, BIN_NM)
    idx = np.digitize(w, edges) - 1
    u = np.unique(idx)
    r = np.column_stack([r[:, idx == k].mean(1) for k in u])
    w = np.array([w[idx == k].mean() for k in u])
    near = lambda nm: int(np.argmin(abs(w - nm)))
    ndvi = (r[:, near(800)] - r[:, near(680)]) / (r[:, near(800)] + r[:, near(680)] + 1e-12)
    sr = r[:, near(800)] / (r[:, near(680)] + 1e-12)
    ind = np.column_stack([ndvi, sr, r.mean(1), r.std(1), r.max(1) - r.min(1),
                           np.percentile(r, 25, axis=1), np.percentile(r, 75, axis=1),
                           kurtosis(r, axis=1), skew(r, axis=1)])
    blocks = [snv(r), ind] + ([META_BLOCK] if with_meta else [])
    return np.hstack(blocks)

# ============================================================
# 3. MODELOS
# ============================================================
class Prep:
    def __init__(self, n_sel=300, n_pc=30, seed=42):
        self.n_sel, self.n_pc, self.seed = n_sel, n_pc, seed
    def fit(self, X, y):
        self.sc1 = StandardScaler().fit(X); Xs = self.sc1.transform(X)
        et = ExtraTreesRegressor(300, max_features=0.33, random_state=self.seed, n_jobs=-1).fit(Xs, y)
        k = min(self.n_sel, X.shape[1])
        self.sel = np.sort(np.argsort(et.feature_importances_)[::-1][:k])
        self.pca = PCA(min(self.n_pc, k), random_state=self.seed).fit(Xs[:, self.sel])
        self.sc2 = StandardScaler().fit(self.pca.transform(Xs[:, self.sel]))
        return self
    def sel_tf(self, X): return self.sc1.transform(X)[:, self.sel]
    def tf(self, X): return self.sc2.transform(self.pca.transform(self.sel_tf(X)))

MLP_CONFIGS = [((300, 150, 75, 35), 0.0003, 0.001, 32), ((400, 200, 100, 50), 0.0001, 0.001, 64),
               ((300, 150, 75), 0.0005, 0.003, 32), ((500, 250, 100), 0.0002, 0.001, 128),
               ((200, 100, 50, 25), 0.0005, 0.005, 32)]

def fit_predict(Xtr, ytr, Xte):
    prep = Prep().fit(Xtr, ytr)
    Str, Ste, Ztr, Zte = prep.sel_tf(Xtr), prep.sel_tf(Xte), prep.tf(Xtr), prep.tf(Xte)
    p_et = ExtraTreesRegressor(500, random_state=42, n_jobs=-1).fit(Str, ytr).predict(Ste)
    ml = []
    for s in range(N_SEEDS):
        for i, (a, al, lr, bs) in enumerate(MLP_CONFIGS):
            mdl = MLPRegressor(hidden_layer_sizes=a, activation="relu", solver="adam", alpha=al,
                               learning_rate_init=lr, batch_size=bs, early_stopping=True,
                               validation_fraction=0.15, n_iter_no_change=20, max_iter=2000,
                               random_state=42 + i + 100 * s).fit(Ztr, ytr)
            ml.append(mdl.predict(Zte))
    p_mlp = np.mean(ml, axis=0)
    return {"ET": p_et, "MLP": p_mlp, "ET+MLP": (p_et + p_mlp) / 2}

def loro_spectral(with_meta, label):
    X = build_features(with_meta)
    P = {k: np.zeros(n) for k in ("ET", "MLP", "ET+MLP")}
    t0 = time.time()
    for g in uniq:
        te, tr = groups == g, groups != g
        out = fit_predict(X[tr], y[tr], X[te])
        for k in P:
            P[k][te] = out[k]
        print(f"  [{label}] fold {g} ok ({time.time() - t0:.0f}s)")
    return P

# ============================================================
# 4. BASELINES SÓ DE METADADOS
# ============================================================
def meta_only_et():
    p = np.zeros(n)
    for g in uniq:
        te, tr = groups == g, groups != g
        p[te] = ExtraTreesRegressor(500, min_samples_leaf=5, random_state=42,
                                    n_jobs=-1).fit(META_BLOCK[tr], y[tr]).predict(META_BLOCK[te])
    return p

def meta_lookup():
    """Média de Ψw por (mês x manejo) calculada só no treino de cada fold."""
    p = np.zeros(n)
    for g in uniq:
        te, tr = groups == g, groups != g
        d = pd.DataFrame({"m": mo[tr], "g": mg[tr], "y": y[tr]})
        cell = d.groupby(["m", "g"])["y"].mean()
        fb = d.groupby("g")["y"].mean()
        for i in np.where(te)[0]:
            p[i] = cell.get((mo[i], mg[i]), fb[mg[i]])
    return p

# ============================================================
# 5. EXECUÇÃO
# ============================================================
print("\n=== Espectro + metadados ===")
P_full = loro_spectral(True, "espectro+meta")
print("\n=== Só espectro ===")
P_spec = loro_spectral(False, "só espectro")
print("\n=== Só metadados ===")
p_meta_et = meta_only_et()
p_meta_lk = meta_lookup()

MODELS = {
    "Espectro+meta: ET+MLP": P_full["ET+MLP"],
    "Espectro+meta: MLP": P_full["MLP"],
    "Espectro+meta: ET": P_full["ET"],
    "Só espectro: ET+MLP": P_spec["ET+MLP"],
    "Só espectro: MLP": P_spec["MLP"],
    "Só espectro: ET": P_spec["ET"],
    "Só metadados: ET": p_meta_et,
    "Só metadados: média mês x manejo": p_meta_lk,
}

# ============================================================
# 6. MÉTRICAS
# ============================================================
def rmse(yt, yp): return np.sqrt(mean_squared_error(yt, yp))

rows, fold_rows = [], []
for name, p in MODELS.items():
    fr = [r2_score(y[groups == g], p[groups == g]) for g in uniq]
    for g, v in zip(uniq, fr):
        fold_rows.append({"Modelo": name, "Replicata_teste": g, "R2": v})
    row = {"Modelo": name, "R2_folds_mean": np.mean(fr), "R2_folds_std": np.std(fr),
           "R2_Rep4": fr[list(uniq).index(4)] if 4 in uniq else np.nan,
           "R2_global_pooled": r2_score(y, p), "RMSE_global": rmse(y, p),
           "RPD_global": np.std(y, ddof=1) / rmse(y, p)}
    for mname in ("Irrigado", "Sequeiro"):
        m = masks[mname]
        row[f"R2_{mname}"] = r2_score(y[m], p[m])
        row[f"RMSE_{mname}"] = rmse(y[m], p[m])
    rows.append(row)
res = pd.DataFrame(rows)
folds = pd.DataFrame(fold_rows).pivot(index="Modelo", columns="Replicata_teste", values="R2")

print("\n" + "=" * 100 + "\nDESEMPENHO (R² por manejo calculado com as predições out-of-fold agrupadas)\n" + "=" * 100)
print(res.round(4).to_string(index=False))
print(f"\nSD(y): irrigado={y[MG].std(ddof=1):.3f} | sequeiro={y[~MG].std(ddof=1):.3f} "
      f"| global={y.std(ddof=1):.3f}   (RMSE > SD(y) => pior que prever a média do grupo)")
print("\nR² por fold (replicata de teste):")
print(folds.round(3).to_string())

# ============================================================
# 7. BOOTSTRAP EM BLOCOS (réplica x ano x mês)
# ============================================================
cl = pd.factorize(pd.Series(list(zip(groups, df['Year'], df['Month']))))[0]
cl_idx = [np.where(cl == c)[0] for c in range(cl.max() + 1)]

def boot(pa, pb=None, mask=None, B=N_BOOT, seed=0):
    rng = np.random.default_rng(seed)
    blocks = [ix[mask[ix]] for ix in cl_idx] if mask is not None else cl_idx
    blocks = [b for b in blocks if len(b) > 1]
    ra, dd = [], []
    for _ in range(B):
        pick = rng.integers(0, len(blocks), len(blocks))
        ix = np.concatenate([blocks[i] for i in pick])
        a = r2_score(y[ix], pa[ix]); ra.append(a)
        if pb is not None:
            dd.append(a - r2_score(y[ix], pb[ix]))
    ci = np.percentile(ra, [2.5, 97.5])
    di = np.percentile(dd, [2.5, 97.5]) if dd else (np.nan, np.nan)
    return ci, di

print("\n" + "=" * 100 + "\nIC 95% (bootstrap em blocos) do R² pooled\n" + "=" * 100)
ci_rows = []
for name in ["Espectro+meta: ET+MLP", "Só espectro: ET+MLP", "Só metadados: ET",
             "Só metadados: média mês x manejo"]:
    for mname, m in masks.items():
        ci, _ = boot(MODELS[name], mask=None if mname == "Global" else m)
        r2v = r2_score(y[m], MODELS[name][m])
        ci_rows.append({"Modelo": name, "Subconjunto": mname, "R2": r2v, "IC_inf": ci[0], "IC_sup": ci[1]})
ci_tab = pd.DataFrame(ci_rows)
print(ci_tab.round(3).to_string(index=False))

print("\n" + "=" * 100 + "\nDIFERENÇAS DE R² (A - B), IC 95% — o IC inclui 0? => diferença não distinguível do ruído\n" + "=" * 100)
PAIRS = [
    ("Espectro+meta: ET+MLP", "Só metadados: ET",
     "O ESPECTRO acrescenta algo além de mês+manejo?"),
    ("Espectro+meta: ET+MLP", "Só metadados: média mês x manejo",
     "Idem, contra a tabela de médias"),
    ("Espectro+meta: ET+MLP", "Só espectro: ET+MLP",
     "Os METADADOS acrescentam algo além do espectro?"),
    ("Só espectro: ET+MLP", "Só metadados: ET",
     "Só espectro vs só metadados"),
]
diff_rows = []
for a, b, pergunta in PAIRS:
    print(f"\n{pergunta}\n  A = {a}\n  B = {b}")
    for mname, m in masks.items():
        _, di = boot(MODELS[a], MODELS[b], mask=None if mname == "Global" else m)
        d = r2_score(y[m], MODELS[a][m]) - r2_score(y[m], MODELS[b][m])
        flag = "" if (di[0] > 0 or di[1] < 0) else "  <-- IC inclui 0"
        print(f"    {mname:9s} dif R2 = {d:+.3f}   IC95% = [{di[0]:+.3f}, {di[1]:+.3f}]{flag}")
        diff_rows.append({"A": a, "B": b, "Subconjunto": mname, "dif_R2": d,
                          "IC_inf": di[0], "IC_sup": di[1]})
diff_tab = pd.DataFrame(diff_rows)

# ============================================================
# 8. FIGURA
# ============================================================
fig, ax = plt.subplots(1, 2, figsize=(14, 5))
names = [("Espectro+meta: ET+MLP", "#2f3e4e"), ("Só espectro: ET+MLP", "#5aa9e0"),
         ("Só metadados: ET", "#f5b650")]
xs = np.arange(3); wbar = 0.26
for j, (nm, col) in enumerate(names):
    vals, lo, hi = [], [], []
    for mname in ("Global", "Irrigado", "Sequeiro"):
        r_ = ci_tab[(ci_tab.Modelo == nm) & (ci_tab.Subconjunto == mname)].iloc[0]
        vals.append(r_.R2); lo.append(r_.R2 - r_.IC_inf); hi.append(r_.IC_sup - r_.R2)
    ax[0].bar(xs + (j - 1) * wbar, vals, wbar, yerr=[lo, hi], capsize=3, color=col, label=nm)
ax[0].axhline(0, color="k", lw=.8)
ax[0].set_xticks(xs); ax[0].set_xticklabels(["Global", "Irrigado", "Sequeiro"])
ax[0].set(ylabel="R² (out-of-fold, IC 95%)", title="Quanto o espectro acrescenta?")
ax[0].legend(fontsize=8)

p_full, p_meta = MODELS["Espectro+meta: ET+MLP"], MODELS["Só metadados: ET"]
ax[1].scatter(y, p_meta, s=12, alpha=.45, color="#f5b650", label=f"Só metadados (R²={r2_score(y, p_meta):.3f})")
ax[1].scatter(y, p_full, s=12, alpha=.45, color="#2f3e4e", label=f"Espectro+meta (R²={r2_score(y, p_full):.3f})")
lim = [y.min() - .2, y.max() + .2]; ax[1].plot(lim, lim, "r--")
ax[1].set(xlabel="Observado (MPa)", ylabel="Predito (MPa)", title="Predições out-of-fold")
ax[1].legend(fontsize=8)
plt.tight_layout(); plt.savefig(f"{OUT_DIR}/fig_v4_baseline_metadados.png", dpi=300); plt.show()

# ============================================================
# 9. SALVAR
# ============================================================
with pd.ExcelWriter(f"{OUT_DIR}/resultados_v4.xlsx") as xw:
    res.to_excel(xw, "desempenho", index=False)
    folds.to_excel(xw, "R2_por_fold")
    ci_tab.to_excel(xw, "IC_R2", index=False)
    diff_tab.to_excel(xw, "diferencas_IC", index=False)
    pd.DataFrame({"replicate": groups, "manejo": df['Sistema_Manejo'].values, "month": mo,
                  "y_true": y, **{k: v for k, v in MODELS.items()}}).to_excel(xw, "predicoes_oof", index=False)
print(f"\nConcluído. Arquivos em: {OUT_DIR}")
```

```python
# -*- coding: utf-8 -*-
# @title EAAI pipeline - V5: o resultado do V4 se mantém quando uma DATA inteira fica fora do treino?
"""
EAAI pipeline - V5: o resultado do V4 se mantém quando uma DATA inteira fica fora do treino?

Modelos comparados (todos com validação Leave-One-Replicate-Out):
  1. Espectro + metadados (mês + manejo)   : ET, ensemble de 5 MLPs, ET+MLP
  2. Só espectro                           : ET, ensemble de 5 MLPs, ET+MLP
  3. Só metadados (mês + manejo)           : Extra Trees (min_samples_leaf=5)
  4. Só metadados, tabela de médias        : média de Ψw por (mês x manejo) do treino

Saídas:
  - R² por fold, global e dentro de cada manejo (Irrigado / Sequeiro)
  - IC 95% por bootstrap em blocos (réplica x ano x mês) do R² e das DIFERENÇAS
    entre modelos (global e dentro de cada manejo)
  - figura e Excel em OUT_DIR
Configuração espectral: 406-900 nm, janelas de 5 nm, SEM o ano como preditor.
Não testado: rode e ajuste se necessário.
"""
import os, time, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import skew, kurtosis
from sklearn.decomposition import PCA
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.metrics import r2_score, mean_squared_error
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler

warnings.filterwarnings("ignore")
np.random.seed(42)
plt.rcParams.update({"figure.dpi": 110, "axes.grid": True, "grid.alpha": .25})

# ============================================================
# CONFIG
# ============================================================
BASE_PATH = '/content/drive/MyDrive/Tese/propostas_de_artigos/proposta_artigo_043'
DATA_PATH = os.path.join(BASE_PATH, '/data/dataset_unificado_043.xlsx')
OUT_DIR = os.path.join(BASE_PATH, "/resultado/resultados_artigo_043_V5_periodo")
os.makedirs(OUT_DIR, exist_ok=True)

WL_MIN, WL_MAX, BIN_NM = 406.0, 900.0, 5
N_SEEDS = 1   # 12 folds: 1 semente por arquitetura para ficar rápido
N_BOOT = 1000

# ============================================================
# 1. LEITURA
# ============================================================
df = pd.read_excel(DATA_PATH)
spec_cols = [c for c in df.columns if str(c).endswith('nm')]
wls_all = np.array([float(str(c).replace('nm', '')) for c in spec_cols])
o = np.argsort(wls_all); spec_cols = [spec_cols[i] for i in o]; wls_all = wls_all[o]
R = df[spec_cols].apply(pd.to_numeric, errors="coerce").interpolate(axis=1, limit_direction="both")
y_all = pd.to_numeric(df['Water_Potential'], errors="coerce")
ok = R.notna().all(axis=1) & y_all.notna() & df['Replicate'].notna()
df, R, y_all = df[ok].reset_index(drop=True), R[ok].reset_index(drop=True), y_all[ok].reset_index(drop=True)
y = y_all.values.astype(float)
groups = df['Replicate'].values.astype(int)
raw_all = R.values.astype(float)
uniq, n = np.unique(groups), len(y)

# ---- Modo de validação -------------------------------------------------
# 'replicate' = LORO (como no artigo; datas já vistas no treino)
# 'period'    = Leave-One-Period-Out (ano x mês): data inteira fica fora do treino
# 'year'      = Leave-One-Year-Out (3 folds)
CV_MODE = 'period'
if CV_MODE == 'period':
    codes, uniques = pd.factorize(pd.Series(list(zip(df['Year'].values, df['Month'].values))))
    groups = codes + 1
    print("Períodos (fold -> (ano, mês)):", {i + 1: tuple(int(v) for v in u) for i, u in enumerate(uniques)})
elif CV_MODE == 'year':
    groups = df['Year'].values.astype(int)
uniq = np.unique(groups)
print(f"CV_MODE={CV_MODE} | {len(uniq)} folds | amostras por fold: {dict(pd.Series(groups).value_counts().sort_index())}")

mo = df['Month'].values.astype(float)
mg = (df['Sistema_Manejo'] == 'Irrigado').astype(float).values
MG = mg == 1
masks = {"Global": np.ones(n, bool), "Irrigado": MG, "Sequeiro": ~MG}
print(f"Amostras: {n} | irrigado={MG.sum()} | sequeiro={(~MG).sum()} | replicatas={list(uniq)}")

# ============================================================
# 2. FEATURES ESPECTRAIS
# ============================================================
def snv(M):
    return (M - M.mean(1, keepdims=True)) / (M.std(1, keepdims=True) + 1e-12)

META_BLOCK = np.column_stack([np.sin(2*np.pi*mo/12), np.cos(2*np.pi*mo/12),
                              np.sin(4*np.pi*mo/12), np.cos(4*np.pi*mo/12), mg])

def build_features(with_meta):
    m = (wls_all >= WL_MIN) & (wls_all <= WL_MAX)
    r, w = raw_all[:, m], wls_all[m]
    edges = np.arange(w.min(), w.max() + BIN_NM, BIN_NM)
    idx = np.digitize(w, edges) - 1
    u = np.unique(idx)
    r = np.column_stack([r[:, idx == k].mean(1) for k in u])
    w = np.array([w[idx == k].mean() for k in u])
    near = lambda nm: int(np.argmin(abs(w - nm)))
    ndvi = (r[:, near(800)] - r[:, near(680)]) / (r[:, near(800)] + r[:, near(680)] + 1e-12)
    sr = r[:, near(800)] / (r[:, near(680)] + 1e-12)
    ind = np.column_stack([ndvi, sr, r.mean(1), r.std(1), r.max(1) - r.min(1),
                           np.percentile(r, 25, axis=1), np.percentile(r, 75, axis=1),
                           kurtosis(r, axis=1), skew(r, axis=1)])
    blocks = [snv(r), ind] + ([META_BLOCK] if with_meta else [])
    return np.hstack(blocks)

# ============================================================
# 3. MODELOS
# ============================================================
class Prep:
    def __init__(self, n_sel=300, n_pc=30, seed=42):
        self.n_sel, self.n_pc, self.seed = n_sel, n_pc, seed
    def fit(self, X, y):
        self.sc1 = StandardScaler().fit(X); Xs = self.sc1.transform(X)
        et = ExtraTreesRegressor(300, max_features=0.33, random_state=self.seed, n_jobs=-1).fit(Xs, y)
        k = min(self.n_sel, X.shape[1])
        self.sel = np.sort(np.argsort(et.feature_importances_)[::-1][:k])
        self.pca = PCA(min(self.n_pc, k), random_state=self.seed).fit(Xs[:, self.sel])
        self.sc2 = StandardScaler().fit(self.pca.transform(Xs[:, self.sel]))
        return self
    def sel_tf(self, X): return self.sc1.transform(X)[:, self.sel]
    def tf(self, X): return self.sc2.transform(self.pca.transform(self.sel_tf(X)))

MLP_CONFIGS = [((300, 150, 75, 35), 0.0003, 0.001, 32), ((400, 200, 100, 50), 0.0001, 0.001, 64),
               ((300, 150, 75), 0.0005, 0.003, 32), ((500, 250, 100), 0.0002, 0.001, 128),
               ((200, 100, 50, 25), 0.0005, 0.005, 32)]

def fit_predict(Xtr, ytr, Xte):
    prep = Prep().fit(Xtr, ytr)
    Str, Ste, Ztr, Zte = prep.sel_tf(Xtr), prep.sel_tf(Xte), prep.tf(Xtr), prep.tf(Xte)
    p_et = ExtraTreesRegressor(500, random_state=42, n_jobs=-1).fit(Str, ytr).predict(Ste)
    ml = []
    for s in range(N_SEEDS):
        for i, (a, al, lr, bs) in enumerate(MLP_CONFIGS):
            mdl = MLPRegressor(hidden_layer_sizes=a, activation="relu", solver="adam", alpha=al,
                               learning_rate_init=lr, batch_size=bs, early_stopping=True,
                               validation_fraction=0.15, n_iter_no_change=20, max_iter=2000,
                               random_state=42 + i + 100 * s).fit(Ztr, ytr)
            ml.append(mdl.predict(Zte))
    p_mlp = np.mean(ml, axis=0)
    return {"ET": p_et, "MLP": p_mlp, "ET+MLP": (p_et + p_mlp) / 2}

def loro_spectral(with_meta, label):
    X = build_features(with_meta)
    P = {k: np.zeros(n) for k in ("ET", "MLP", "ET+MLP")}
    t0 = time.time()
    for g in uniq:
        te, tr = groups == g, groups != g
        out = fit_predict(X[tr], y[tr], X[te])
        for k in P:
            P[k][te] = out[k]
        print(f"  [{label}] fold {g} ok ({time.time() - t0:.0f}s)")
    return P

# ============================================================
# 4. BASELINES SÓ DE METADADOS
# ============================================================
def meta_only_et():
    p = np.zeros(n)
    for g in uniq:
        te, tr = groups == g, groups != g
        p[te] = ExtraTreesRegressor(500, min_samples_leaf=5, random_state=42,
                                    n_jobs=-1).fit(META_BLOCK[tr], y[tr]).predict(META_BLOCK[te])
    return p

def meta_lookup():
    """Média de Ψw por (mês x manejo) calculada só no treino de cada fold."""
    p = np.zeros(n)
    for g in uniq:
        te, tr = groups == g, groups != g
        d = pd.DataFrame({"m": mo[tr], "g": mg[tr], "y": y[tr]})
        cell = d.groupby(["m", "g"])["y"].mean()
        fb = d.groupby("g")["y"].mean()
        for i in np.where(te)[0]:
            p[i] = cell.get((mo[i], mg[i]), fb[mg[i]])
    return p

# ============================================================
# 5. EXECUÇÃO
# ============================================================
print("\n=== Espectro + metadados ===")
P_full = loro_spectral(True, "espectro+meta")
print("\n=== Só espectro ===")
P_spec = loro_spectral(False, "só espectro")
print("\n=== Só metadados ===")
p_meta_et = meta_only_et()
p_meta_lk = meta_lookup()

MODELS = {
    "Espectro+meta: ET+MLP": P_full["ET+MLP"],
    "Espectro+meta: MLP": P_full["MLP"],
    "Espectro+meta: ET": P_full["ET"],
    "Só espectro: ET+MLP": P_spec["ET+MLP"],
    "Só espectro: MLP": P_spec["MLP"],
    "Só espectro: ET": P_spec["ET"],
    "Só metadados: ET": p_meta_et,
    "Só metadados: média mês x manejo": p_meta_lk,
}

# ============================================================
# 6. MÉTRICAS
# ============================================================
def rmse(yt, yp): return np.sqrt(mean_squared_error(yt, yp))

rows, fold_rows = [], []
for name, p in MODELS.items():
    fr = [r2_score(y[groups == g], p[groups == g]) for g in uniq]
    for g, v in zip(uniq, fr):
        fold_rows.append({"Modelo": name, "Replicata_teste": g, "R2": v})
    row = {"Modelo": name, "R2_folds_mean": np.mean(fr), "R2_folds_std": np.std(fr),
           "R2_pior_fold": np.min(fr),
           "R2_global_pooled": r2_score(y, p), "RMSE_global": rmse(y, p),
           "RPD_global": np.std(y, ddof=1) / rmse(y, p)}
    for mname in ("Irrigado", "Sequeiro"):
        m = masks[mname]
        row[f"R2_{mname}"] = r2_score(y[m], p[m])
        row[f"RMSE_{mname}"] = rmse(y[m], p[m])
    rows.append(row)
res = pd.DataFrame(rows)
folds = pd.DataFrame(fold_rows).pivot(index="Modelo", columns="Replicata_teste", values="R2")

print("\n" + "=" * 100 + "\nDESEMPENHO (R² por manejo calculado com as predições out-of-fold agrupadas)\n" + "=" * 100)
print(res.round(4).to_string(index=False))
print(f"\nSD(y): irrigado={y[MG].std(ddof=1):.3f} | sequeiro={y[~MG].std(ddof=1):.3f} "
      f"| global={y.std(ddof=1):.3f}   (RMSE > SD(y) => pior que prever a média do grupo)")
print("\nR² por fold (replicata de teste):")
print(folds.round(3).to_string())

# ============================================================
# 7. BOOTSTRAP EM BLOCOS (réplica x ano x mês)
# ============================================================
cl = pd.factorize(pd.Series(list(zip(groups, df['Year'], df['Month']))))[0]
cl_idx = [np.where(cl == c)[0] for c in range(cl.max() + 1)]

def boot(pa, pb=None, mask=None, B=N_BOOT, seed=0):
    rng = np.random.default_rng(seed)
    blocks = [ix[mask[ix]] for ix in cl_idx] if mask is not None else cl_idx
    blocks = [b for b in blocks if len(b) > 1]
    ra, dd = [], []
    for _ in range(B):
        pick = rng.integers(0, len(blocks), len(blocks))
        ix = np.concatenate([blocks[i] for i in pick])
        a = r2_score(y[ix], pa[ix]); ra.append(a)
        if pb is not None:
            dd.append(a - r2_score(y[ix], pb[ix]))
    ci = np.percentile(ra, [2.5, 97.5])
    di = np.percentile(dd, [2.5, 97.5]) if dd else (np.nan, np.nan)
    return ci, di

print("\n" + "=" * 100 + "\nIC 95% (bootstrap em blocos) do R² pooled\n" + "=" * 100)
ci_rows = []
for name in ["Espectro+meta: ET+MLP", "Só espectro: ET+MLP", "Só metadados: ET",
             "Só metadados: média mês x manejo"]:
    for mname, m in masks.items():
        ci, _ = boot(MODELS[name], mask=None if mname == "Global" else m)
        r2v = r2_score(y[m], MODELS[name][m])
        ci_rows.append({"Modelo": name, "Subconjunto": mname, "R2": r2v, "IC_inf": ci[0], "IC_sup": ci[1]})
ci_tab = pd.DataFrame(ci_rows)
print(ci_tab.round(3).to_string(index=False))

print("\n" + "=" * 100 + "\nDIFERENÇAS DE R² (A - B), IC 95% — o IC inclui 0? => diferença não distinguível do ruído\n" + "=" * 100)
PAIRS = [
    ("Espectro+meta: ET+MLP", "Só metadados: ET",
     "O ESPECTRO acrescenta algo além de mês+manejo?"),
    ("Espectro+meta: ET+MLP", "Só metadados: média mês x manejo",
     "Idem, contra a tabela de médias"),
    ("Espectro+meta: ET+MLP", "Só espectro: ET+MLP",
     "Os METADADOS acrescentam algo além do espectro?"),
    ("Só espectro: ET+MLP", "Só metadados: ET",
     "Só espectro vs só metadados"),
]
diff_rows = []
for a, b, pergunta in PAIRS:
    print(f"\n{pergunta}\n  A = {a}\n  B = {b}")
    for mname, m in masks.items():
        _, di = boot(MODELS[a], MODELS[b], mask=None if mname == "Global" else m)
        d = r2_score(y[m], MODELS[a][m]) - r2_score(y[m], MODELS[b][m])
        flag = "" if (di[0] > 0 or di[1] < 0) else "  <-- IC inclui 0"
        print(f"    {mname:9s} dif R2 = {d:+.3f}   IC95% = [{di[0]:+.3f}, {di[1]:+.3f}]{flag}")
        diff_rows.append({"A": a, "B": b, "Subconjunto": mname, "dif_R2": d,
                          "IC_inf": di[0], "IC_sup": di[1]})
diff_tab = pd.DataFrame(diff_rows)

# ============================================================
# 8. FIGURA
# ============================================================
fig, ax = plt.subplots(1, 2, figsize=(14, 5))
names = [("Espectro+meta: ET+MLP", "#2f3e4e"), ("Só espectro: ET+MLP", "#5aa9e0"),
         ("Só metadados: ET", "#f5b650")]
xs = np.arange(3); wbar = 0.26
for j, (nm, col) in enumerate(names):
    vals, lo, hi = [], [], []
    for mname in ("Global", "Irrigado", "Sequeiro"):
        r_ = ci_tab[(ci_tab.Modelo == nm) & (ci_tab.Subconjunto == mname)].iloc[0]
        vals.append(r_.R2); lo.append(r_.R2 - r_.IC_inf); hi.append(r_.IC_sup - r_.R2)
    ax[0].bar(xs + (j - 1) * wbar, vals, wbar, yerr=[lo, hi], capsize=3, color=col, label=nm)
ax[0].axhline(0, color="k", lw=.8)
ax[0].set_xticks(xs); ax[0].set_xticklabels(["Global", "Irrigado", "Sequeiro"])
ax[0].set(ylabel="R² (out-of-fold, IC 95%)", title="Quanto o espectro acrescenta?")
ax[0].legend(fontsize=8)

p_full, p_meta = MODELS["Espectro+meta: ET+MLP"], MODELS["Só metadados: ET"]
ax[1].scatter(y, p_meta, s=12, alpha=.45, color="#f5b650", label=f"Só metadados (R²={r2_score(y, p_meta):.3f})")
ax[1].scatter(y, p_full, s=12, alpha=.45, color="#2f3e4e", label=f"Espectro+meta (R²={r2_score(y, p_full):.3f})")
lim = [y.min() - .2, y.max() + .2]; ax[1].plot(lim, lim, "r--")
ax[1].set(xlabel="Observado (MPa)", ylabel="Predito (MPa)", title="Predições out-of-fold")
ax[1].legend(fontsize=8)
plt.tight_layout(); plt.savefig(f"{OUT_DIR}/fig_v5_periodo.png", dpi=300); plt.show()

# ============================================================
# 9. SALVAR
# ============================================================
with pd.ExcelWriter(f"{OUT_DIR}/resultados_v5.xlsx") as xw:
    res.to_excel(xw, "desempenho", index=False)
    folds.to_excel(xw, "R2_por_fold")
    ci_tab.to_excel(xw, "IC_R2", index=False)
    diff_tab.to_excel(xw, "diferencas_IC", index=False)
    pd.DataFrame({"replicate": groups, "manejo": df['Sistema_Manejo'].values, "month": mo,
                  "y_true": y, **{k: v for k, v in MODELS.items()}}).to_excel(xw, "predicoes_oof", index=False)
print(f"\nConcluído. Arquivos em: {OUT_DIR}")
```

```python
# -*- coding: utf-8 -*-
# @title EAAI pipeline - V4: How much does the SPECTRUM add beyond month + management?
"""
EAAI pipeline - V4 (adapted for 4-panel figure in English)
"""
import os, time, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import skew, kurtosis
from sklearn.decomposition import PCA
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.metrics import r2_score, mean_squared_error
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler

warnings.filterwarnings("ignore")
np.random.seed(42)
plt.rcParams.update({"figure.dpi": 110, "axes.grid": True, "grid.alpha": .25})

# ============================================================
# CONFIG
# ============================================================
BASE_PATH = '/content/drive/MyDrive/Tese/propostas_de_artigos/proposta_artigo_043'
DATA_PATH = os.path.join(BASE_PATH, 'data/dataset_unificado_043.xlsx')
OUT_DIR = os.path.join(BASE_PATH, "resultado/resultados_artigo_043_V4")
os.makedirs(OUT_DIR, exist_ok=True)

WL_MIN, WL_MAX, BIN_NM = 406.0, 900.0, 5
N_SEEDS = 3
N_BOOT = 1000

# ============================================================
# 1. LEITURA
# ============================================================
df = pd.read_excel(DATA_PATH)
spec_cols = [c for c in df.columns if str(c).endswith('nm')]
wls_all = np.array([float(str(c).replace('nm', '')) for c in spec_cols])
o = np.argsort(wls_all); spec_cols = [spec_cols[i] for i in o]; wls_all = wls_all[o]
R = df[spec_cols].apply(pd.to_numeric, errors="coerce").interpolate(axis=1, limit_direction="both")
y_all = pd.to_numeric(df['Water_Potential'], errors="coerce")
ok = R.notna().all(axis=1) & y_all.notna() & df['Replicate'].notna()
df, R, y_all = df[ok].reset_index(drop=True), R[ok].reset_index(drop=True), y_all[ok].reset_index(drop=True)
y = y_all.values.astype(float)
groups = df['Replicate'].values.astype(int)
raw_all = R.values.astype(float)
uniq, n = np.unique(groups), len(y)

mo = df['Month'].values.astype(float)
mg = (df['Sistema_Manejo'] == 'Irrigado').astype(float).values
MG = mg == 1
masks = {"Global": np.ones(n, bool), "Irrigated": MG, "Rainfed": ~MG}
print(f"Samples: {n} | irrigated={MG.sum()} | rainfed={(~MG).sum()} | replicates={list(uniq)}")

# ============================================================
# 2. FEATURES
# ============================================================
def snv(M):
    return (M - M.mean(1, keepdims=True)) / (M.std(1, keepdims=True) + 1e-12)

META_BLOCK = np.column_stack([np.sin(2*np.pi*mo/12), np.cos(2*np.pi*mo/12),
                              np.sin(4*np.pi*mo/12), np.cos(4*np.pi*mo/12), mg])

def build_features(with_meta):
    m = (wls_all >= WL_MIN) & (wls_all <= WL_MAX)
    r, w = raw_all[:, m], wls_all[m]
    edges = np.arange(w.min(), w.max() + BIN_NM, BIN_NM)
    idx = np.digitize(w, edges) - 1
    u = np.unique(idx)
    r = np.column_stack([r[:, idx == k].mean(1) for k in u])
    w = np.array([w[idx == k].mean() for k in u])
    near = lambda nm: int(np.argmin(abs(w - nm)))
    ndvi = (r[:, near(800)] - r[:, near(680)]) / (r[:, near(800)] + r[:, near(680)] + 1e-12)
    sr = r[:, near(800)] / (r[:, near(680)] + 1e-12)
    ind = np.column_stack([ndvi, sr, r.mean(1), r.std(1), r.max(1) - r.min(1),
                           np.percentile(r, 25, axis=1), np.percentile(r, 75, axis=1),
                           kurtosis(r, axis=1), skew(r, axis=1)])
    blocks = [snv(r), ind] + ([META_BLOCK] if with_meta else [])
    return np.hstack(blocks)

# ============================================================
# 3. MODELS
# ============================================================
class Prep:
    def __init__(self, n_sel=300, n_pc=30, seed=42):
        self.n_sel, self.n_pc, self.seed = n_sel, n_pc, seed
    def fit(self, X, y):
        self.sc1 = StandardScaler().fit(X); Xs = self.sc1.transform(X)
        et = ExtraTreesRegressor(300, max_features=0.33, random_state=self.seed, n_jobs=-1).fit(Xs, y)
        k = min(self.n_sel, X.shape[1])
        self.sel = np.sort(np.argsort(et.feature_importances_)[::-1][:k])
        self.pca = PCA(min(self.n_pc, k), random_state=self.seed).fit(Xs[:, self.sel])
        self.sc2 = StandardScaler().fit(self.pca.transform(Xs[:, self.sel]))
        return self
    def sel_tf(self, X): return self.sc1.transform(X)[:, self.sel]
    def tf(self, X): return self.sc2.transform(self.pca.transform(self.sel_tf(X)))

MLP_CONFIGS = [((300, 150, 75, 35), 0.0003, 0.001, 32), ((400, 200, 100, 50), 0.0001, 0.001, 64),
               ((300, 150, 75), 0.0005, 0.003, 32), ((500, 250, 100), 0.0002, 0.001, 128),
               ((200, 100, 50, 25), 0.0005, 0.005, 32)]

def fit_predict(Xtr, ytr, Xte):
    prep = Prep().fit(Xtr, ytr)
    Str, Ste, Ztr, Zte = prep.sel_tf(Xtr), prep.sel_tf(Xte), prep.tf(Xtr), prep.tf(Xte)
    p_et = ExtraTreesRegressor(500, random_state=42, n_jobs=-1).fit(Str, ytr).predict(Ste)
    ml = []
    for s in range(N_SEEDS):
        for i, (a, al, lr, bs) in enumerate(MLP_CONFIGS):
            mdl = MLPRegressor(hidden_layer_sizes=a, activation="relu", solver="adam", alpha=al,
                               learning_rate_init=lr, batch_size=bs, early_stopping=True,
                               validation_fraction=0.15, n_iter_no_change=20, max_iter=2000,
                               random_state=42 + i + 100 * s).fit(Ztr, ytr)
            ml.append(mdl.predict(Zte))
    p_mlp = np.mean(ml, axis=0)
    return {"ET": p_et, "MLP": p_mlp, "ET+MLP": (p_et + p_mlp) / 2}

def loro_spectral(with_meta, label):
    X = build_features(with_meta)
    P = {k: np.zeros(n) for k in ("ET", "MLP", "ET+MLP")}
    t0 = time.time()
    for g in uniq:
        te, tr = groups == g, groups != g
        out = fit_predict(X[tr], y[tr], X[te])
        for k in P:
            P[k][te] = out[k]
        print(f"  [{label}] fold {g} ok ({time.time() - t0:.0f}s)")
    return P

# ============================================================
# 4. META-ONLY BASELINES
# ============================================================
def meta_only_et():
    p = np.zeros(n)
    for g in uniq:
        te, tr = groups == g, groups != g
        p[te] = ExtraTreesRegressor(500, min_samples_leaf=5, random_state=42,
                                    n_jobs=-1).fit(META_BLOCK[tr], y[tr]).predict(META_BLOCK[te])
    return p

def meta_lookup():
    p = np.zeros(n)
    for g in uniq:
        te, tr = groups == g, groups != g
        d = pd.DataFrame({"m": mo[tr], "g": mg[tr], "y": y[tr]})
        cell = d.groupby(["m", "g"])["y"].mean()
        fb = d.groupby("g")["y"].mean()
        for i in np.where(te)[0]:
            p[i] = cell.get((mo[i], mg[i]), fb[mg[i]])
    return p

# ============================================================
# 5. EXECUÇÃO
# ============================================================
print("\n=== Spectrum + metadata ===")
P_full = loro_spectral(True, "spectrum+meta")
print("\n=== Spectrum only ===")
P_spec = loro_spectral(False, "spectrum only")
print("\n=== Metadata only ===")
p_meta_et = meta_only_et()
p_meta_lk = meta_lookup()

MODELS = {
    "Spectrum+meta: ET+MLP": P_full["ET+MLP"],
    "Spectrum+meta: MLP": P_full["MLP"],
    "Spectrum+meta: ET": P_full["ET"],
    "Spectrum only: ET+MLP": P_spec["ET+MLP"],
    "Spectrum only: MLP": P_spec["MLP"],
    "Spectrum only: ET": P_spec["ET"],
    "Metadata only: ET": p_meta_et,
    "Metadata only: mean (month x management)": p_meta_lk,
}

# ============================================================
# 6. MÉTRICAS
# ============================================================
def rmse(yt, yp): return np.sqrt(mean_squared_error(yt, yp))

rows, fold_rows = [], []
for name, p in MODELS.items():
    fr = [r2_score(y[groups == g], p[groups == g]) for g in uniq]
    for g, v in zip(uniq, fr):
        fold_rows.append({"Model": name, "Test_replicate": g, "R2": v})
    row = {"Model": name, "R2_folds_mean": np.mean(fr), "R2_folds_std": np.std(fr),
           "R2_Rep4": fr[list(uniq).index(4)] if 4 in uniq else np.nan,
           "R2_global_pooled": r2_score(y, p), "RMSE_global": rmse(y, p),
           "RPD_global": np.std(y, ddof=1) / rmse(y, p)}
    for mname in ("Irrigated", "Rainfed"):
        m = masks[mname]
        row[f"R2_{mname}"] = r2_score(y[m], p[m])
        row[f"RMSE_{mname}"] = rmse(y[m], p[m])
    rows.append(row)
res = pd.DataFrame(rows)
folds = pd.DataFrame(fold_rows).pivot(index="Model", columns="Test_replicate", values="R2")

print("\n" + "=" * 100 + "\nPERFORMANCE (R² per management computed with pooled out-of-fold predictions)\n" + "=" * 100)
print(res.round(4).to_string(index=False))
print(f"\nSD(y): irrigated={y[MG].std(ddof=1):.3f} | rainfed={y[~MG].std(ddof=1):.3f} "
      f"| global={y.std(ddof=1):.3f}   (RMSE > SD(y) => worse than predicting the group mean)")

# ============================================================
# 7. BLOCK BOOTSTRAP
# ============================================================
cl = pd.factorize(pd.Series(list(zip(groups, df['Year'], df['Month']))))[0]
cl_idx = [np.where(cl == c)[0] for c in range(cl.max() + 1)]

def boot(pa, pb=None, mask=None, B=N_BOOT, seed=0):
    rng = np.random.default_rng(seed)
    blocks = [ix[mask[ix]] for ix in cl_idx] if mask is not None else cl_idx
    blocks = [b for b in blocks if len(b) > 1]
    ra, dd = [], []
    for _ in range(B):
        pick = rng.integers(0, len(blocks), len(blocks))
        ix = np.concatenate([blocks[i] for i in pick])
        a = r2_score(y[ix], pa[ix]); ra.append(a)
        if pb is not None:
            dd.append(a - r2_score(y[ix], pb[ix]))
    ci = np.percentile(ra, [2.5, 97.5])
    di = np.percentile(dd, [2.5, 97.5]) if dd else (np.nan, np.nan)
    return ci, di

print("\n" + "=" * 100 + "\n95% CI (block bootstrap) of R² pooled\n" + "=" * 100)
ci_rows = []
for name in ["Spectrum+meta: ET+MLP", "Spectrum only: ET+MLP", "Metadata only: ET",
             "Metadata only: mean (month x management)"]:
    for mname, m in masks.items():
        ci, _ = boot(MODELS[name], mask=None if mname == "Global" else m)
        r2v = r2_score(y[m], MODELS[name][m])
        ci_rows.append({"Model": name, "Subset": mname, "R2": r2v, "CI_low": ci[0], "CI_high": ci[1]})
ci_tab = pd.DataFrame(ci_rows)
print(ci_tab.round(3).to_string(index=False))

# ============================================================
# 8. FIGURA DE 4 PAINÉIS (em inglês)
# ============================================================
# Usar o melhor modelo (Spectrum+meta: ET+MLP) como referência
best_name = "Spectrum+meta: ET+MLP"
best_pred = MODELS[best_name]
residuals = y - best_pred
r2_best = r2_score(y, best_pred)
rmse_best = np.sqrt(mean_squared_error(y, best_pred))
rpd_best = np.std(y, ddof=1) / rmse_best

# Dados para o painel (c): ranking das variantes (ablação)
# Usar os modelos de ablação (Spectrum+meta, Spectrum only, Metadata only)
ablation_names = [
    "Spectrum+meta: ET+MLP",
    "Spectrum+meta: MLP",
    "Spectrum+meta: ET",
    "Spectrum only: ET+MLP",
    "Spectrum only: MLP",
    "Spectrum only: ET",
    "Metadata only: ET",
    "Metadata only: mean (month x management)",
]
ablation_labels = [
    "Spectrum + meta (ET+MLP)",
    "Spectrum + meta (MLP)",
    "Spectrum + meta (ET)",
    "Spectrum only (ET+MLP)",
    "Spectrum only (MLP)",
    "Spectrum only (ET)",
    "Metadata only (ET)",
    "Metadata only (mean)",
]
ablation_r2 = [r2_score(y, MODELS[n]) for n in ablation_names]

# Dados para o painel (d): RMSE por faixa de estresse
bins = [(-np.inf, -4), (-4, -2), (-2, -1), (-1, np.inf)]
bin_labels = ["< -4 MPa", "-4 to -2", "-2 to -1", "> -1 MPa"]
rmse_by_bin = []
for (lo, hi) in bins:
    m = (y > lo) & (y <= hi)
    if m.sum() > 1:
        rmse_by_bin.append(np.sqrt(np.mean(residuals[m] ** 2)))
    else:
        rmse_by_bin.append(np.nan)

# Criar a figura de 4 painéis
fig, ax = plt.subplots(2, 2, figsize=(13, 10))

# (a) Predito vs. Observado
a = ax[0, 0]
a.scatter(y, best_pred, s=14, alpha=.6, color="#2f3e4e")
lim = [min(y.min(), best_pred.min()) - .2, max(y.max(), best_pred.max()) + .2]
a.plot(lim, lim, "r--", label="1:1")
a.set(xlabel="Observed (MPa)", ylabel="Predicted (MPa)",
      title=f"Blend ET+MLP\nR²={r2_best:.4f} | RPD={rpd_best:.2f}")
a.legend()

# (b) Resíduos vs. Predito
a = ax[0, 1]
a.scatter(best_pred, residuals, s=14, alpha=.6, color="#e45756")
a.axhline(0, color="r", ls="--")
a.set(xlabel="Predicted (MPa)", ylabel="Residual (MPa)",
      title=f"Residuals | RMSE={rmse_best:.4f} MPa")

# (c) Ranking das variantes (ablação)
a = ax[1, 0]
# Ordenar por R² (do maior para o menor)
order = np.argsort(ablation_r2)[::-1]
sorted_names = [ablation_labels[i] for i in order]
sorted_r2 = [ablation_r2[i] for i in order]
colors = ["#2f3e4e" if "ET+MLP" in n else "#5aa9e0" for n in sorted_names]
bars = a.barh(range(len(sorted_names)), sorted_r2, color=colors)
for i, v in enumerate(sorted_r2):
    a.text(v + .005, i, f"{v:.4f}", va="center", fontsize=7)
a.set_yticks(range(len(sorted_names)))
a.set_yticklabels(sorted_names, fontsize=7)
a.set_xlabel("R² (mean across folds)")
a.set_title("Ablation: spectrum vs. metadata")
a.set_xlim(0, 1)

# (d) RMSE por faixa de estresse
a = ax[1, 1]
a.bar(bin_labels, rmse_by_bin, color="#e45756")
a.set_ylabel("RMSE (MPa)")
a.set_title("Error by Ψw range")

# Adicionar letras (a), (b), (c), (d)
for lbl, a in zip("abcd", ax.ravel()):
    a.text(-0.08, 1.05, f"({lbl})", transform=a.transAxes, fontsize=13, weight="bold")

plt.tight_layout()
plt.savefig(f"{OUT_DIR}/fig_v4_4panels_english.png", dpi=300)
plt.show()

# ============================================================
# 9. SALVAR
# ============================================================
with pd.ExcelWriter(f"{OUT_DIR}/resultados_v4.xlsx") as xw:
    res.to_excel(xw, "performance", index=False)
    folds.to_excel(xw, "R2_per_fold")
    ci_tab.to_excel(xw, "CI_R2", index=False)
    pd.DataFrame({"replicate": groups, "management": df['Sistema_Manejo'].values, "month": mo,
                  "y_true": y, **{k: v for k, v in MODELS.items()}}).to_excel(xw, "predictions_oof", index=False)
print(f"\nDone. Files saved in: {OUT_DIR}")
```
